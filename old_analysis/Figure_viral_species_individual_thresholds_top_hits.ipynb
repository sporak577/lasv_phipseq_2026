### files / inputs

```python
import pandas as pd

#-- load 
df_SL_rpk = pd.read_csv('./preprocessing/earliest_tp_rpk_avg_value_for_cohort_techreps_071626.csv').set_index('peptide')
df_hbdb_rpk = pd.read_csv('./preprocessing/hbdb_pass_100k_techreps_averaged_071626.csv').set_index('peptide')

df_sl_z = pd.read_csv('z_score_df_SL_over_US_090226.csv').set_index('peptide')
df_hbdb_z = pd.read_csv('z_hbdb_loo_results_090226.csv').set_index('peptide')

peps_metadata = (
pd.read_csv('lassa_library_sequences_species_standardized_070726.csv').rename(columns={'seq_id': 'peptide'})
)
#-- correct wrong genus name
peps_metadata.loc[peps_metadata['genus'] == 'Arenavirus', 'genus'] = 'Mammarenavirus'
```

### calculate peptide-level sensitivity/specificity

```python
Z_THRESH = 3
SENS_MIN = 0.05
SPEC_MIN = 0.99

pass_sl = df_sl_z > Z_THRESH
pass_hbdb = df_hbdb_z > Z_THRESH

peptide_stats = pd.DataFrame({
    'sensitivity': pass_sl.mean(axis=1),
    'specificity': 1 - pass_hbdb.mean(axis=1)
})

peptide_stats = (
    peptide_stats
    .reset_index()
    .rename(columns={'index': 'peptide'})
    .merge(
        peps_metadata[
            [
                'peptide',
                'sequence',
                'species',
                'genus',
                'family',
                'protein_name',
                'accession'
            ]
        ],
        on='peptide',
        how='left'
    )
)
```

### define hits

```python
hits = peptide_stats[
    (peptide_stats['sensitivity'] >= SENS_MIN) &
    (peptide_stats['specificity'] >= SPEC_MIN)
].copy()
```

```python
hits.shape
```

### filtering out peptides we don't want downstream, such as Trypanosomatidae and Konkoviridae taxa

```python
#filtering out taxa I don't want to include downstream, such as Trypanosomatidae and Konkoviridae

TAXON_LEVEL = 'family'
TAXON_TO_REMOVE = ['Trypanosomatidae', 'Konkoviridae']

ACCESSION_TO_REMOVE = ['WAK75154.1', 'ANT96597.1'] #this is an ampicillin resistance protein from an artificial Zika 'vaccine' vector

hits = hits[~hits[TAXON_LEVEL].isin(TAXON_TO_REMOVE)].copy()

hits = hits[~hits['accession'].isin(ACCESSION_TO_REMOVE)]

hits['species'].nunique()
hits.shape
```

### building a long and master dataframe

```python
df_z = pd.concat(
    [df_sl_z, df_hbdb_z],
    axis=1
)

z_long = (
    df_z
    .loc[df_z.index.intersection(hits['peptide'])]
    .reset_index()
    .melt(
        id_vars='peptide',
        var_name='individual',
        value_name='z_score'
    )
)
```

```python
z_long = z_long.merge(
    hits[
        [
            'peptide',
            'sequence',
            'species',
            'genus',
            'family',
            'protein_name'
        ]
    ].drop_duplicates('peptide'),
    on='peptide',
    how='left'
)
```

```python
import numpy as np

z_long['cohort'] = np.where(
    z_long['individual'].str.startswith('HBDB-'),
    'HBDB',
    'SL'
)

z_long['reactive'] = z_long['z_score'] > Z_THRESH
```

### species-level descriptive seroreactivity

```python
species_individual = (
    z_long
    .groupby(
        ['species', 'individual', 'cohort'],
        dropna=False
    )
    .agg(
        n_reactive_peptides=('reactive', 'sum')
    )
    .reset_index()
)

species_individual['seroreactive'] = (
    species_individual['n_reactive_peptides'] > 0
)
```

```python
species_prevalence = (
    species_individual
    .groupby(
        ['species', 'cohort']
    )['seroreactive']
    .mean()
    .reset_index(name='fraction_seroreactive')
)
```

### highest-prevalence species within each genus figure

```python
# ============================================================
# FIGURE 1: TOP SPECIES WITHIN EACH VIRAL FAMILY
# ============================================================

# Add family annotation
species_taxonomy = (
    hits[['species', 'genus', 'family']]
    .drop_duplicates('species')
)

species_prev = (
    species_prevalence
    .pivot(
        index='species',
        columns='cohort',
        values='fraction_seroreactive'
    )
    .fillna(0)
    .reset_index()
    .merge(
        species_taxonomy,
        on='species',
        how='left'
    )
)

# Make sure SL exists
if 'SL' not in species_prev.columns:
    raise ValueError("SL cohort not found in species_prevalence")


# ============================================================
# STEP 1:
# Select the most prevalent species within each FAMILY
# ============================================================

top_species_per_family = (
    species_prev
    .sort_values(
        ['family', 'SL'],
        ascending=[True, False]
    )
    .groupby(
        'family',
        as_index=False
    )
    .head(1)
    .copy()
)


# ============================================================
# STEP 2:
# Take the top N families based on their representative
# species' SL prevalence
# ============================================================

N_FAMILIES = 10

plot_df = (
    top_species_per_family
    .sort_values(
        'SL',
        ascending=False
    )
    .head(N_FAMILIES)
    .sort_values(
        'SL',
        ascending=True
    )
    .copy()
)


# ============================================================
# LABEL
# ============================================================

plot_df['label'] = (
    plot_df['species']
    + '  (' + plot_df['family'] + ')'
)


# ============================================================
# PLOT
# ============================================================

fig, ax = plt.subplots(
    figsize=(9, 6)
)

y = np.arange(len(plot_df))


# ------------------------------------------------------------
# HBDB
# ------------------------------------------------------------

if 'HBDB' in plot_df.columns:

    ax.scatter(
        plot_df['HBDB'],
        y,
        s=65,
        label='US healthy',
        zorder=3
    )


# ------------------------------------------------------------
# SL
# ------------------------------------------------------------

ax.scatter(
    plot_df['SL'],
    y,
    s=65,
    label='SL',
    zorder=3
)


# ------------------------------------------------------------
# Connecting lines
# ------------------------------------------------------------

if 'HBDB' in plot_df.columns:

    for i, (_, row) in enumerate(
        plot_df.iterrows()
    ):

        ax.plot(
            [
                row['HBDB'],
                row['SL']
            ],
            [
                i,
                i
            ],
            linewidth=1,
            alpha=0.5,
            zorder=1
        )


# ------------------------------------------------------------
# Labels
# ------------------------------------------------------------

ax.set_yticks(y)

ax.set_yticklabels(
    plot_df['label']
)


# ------------------------------------------------------------
# Formatting
# ------------------------------------------------------------

ax.set_xlabel(
    'Fraction seroreactive',
    fontsize=13
)

ax.set_ylabel('')

ax.xaxis.set_major_formatter(
    plt.FuncFormatter(
        lambda x, pos: f'{x:.0%}'
    )
)

ax.tick_params(
    axis='both',
    labelsize=11
)

ax.legend(
    frameon=False,
    loc='lower right'
)

ax.set_title(
    'Most prevalent seroreactive species across viral families',
    fontsize=14,
    pad=12
)

sns.despine(ax=ax)

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# FIGURE 1:
# POPULATION SEROREACTIVITY + PEPTIDE-LEVEL RESPONSE DEPTH
#
# Panel A:
#   Fraction seroreactive in SL vs US healthy
#
# Panel B:
#   Number of reactive peptides per seropositive SL individual
#   for each viral species
#
# Species are the top representative species from each family.
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. PREPARE SPECIES-LEVEL SEROREACTIVITY DATA
# ============================================================

# We assume `plot_df` already exists from your previous figure.
# It contains the selected representative species.

selected_species = (
    plot_df['species']
    .tolist()
)


# ============================================================
# 2. PEPTIDE-LEVEL DATA FOR SELECTED SPECIES
# ============================================================

selected_z = (
    z_long[
        z_long['species'].isin(
            selected_species
        )
    ]
    .copy()
)


# ============================================================
# 3. COUNT REACTIVE PEPTIDES PER INDIVIDUAL
#
# A peptide is reactive if z > 3.
#
# Only SL individuals are used for Panel B.
# ============================================================

sl_selected_z = (
    selected_z[
        selected_z['cohort'] == 'SL'
    ]
    .copy()
)

sl_selected_z['reactive'] = (
    sl_selected_z['z_score'] > 3
)


# Count UNIQUE reactive peptides per person/species
peptide_breadth = (
    sl_selected_z[
        sl_selected_z['reactive']
    ]
    .groupby(
        [
            'species',
            'individual'
        ]
    )['peptide']
    .nunique()
    .reset_index(
        name='n_reactive_peptides'
    )
)


# ============================================================
# 4. IDENTIFY SEROPOSITIVE SL INDIVIDUALS
#
# A person is seropositive for a species if they have
# >=1 reactive peptide.
# ============================================================

seropositive_sl = (
    sl_selected_z
    .groupby(
        [
            'species',
            'individual'
        ]
    )['reactive']
    .any()
    .reset_index(
        name='seroreactive'
    )
)

seropositive_sl = (
    seropositive_sl[
        seropositive_sl['seroreactive']
    ]
    .copy()
)


# ============================================================
# 5. MERGE BREADTH
# ============================================================

peptide_breadth = (
    seropositive_sl[
        [
            'species',
            'individual'
        ]
    ]
    .merge(
        peptide_breadth,
        on=[
            'species',
            'individual'
        ],
        how='left'
    )
)

peptide_breadth[
    'n_reactive_peptides'
] = (
    peptide_breadth[
        'n_reactive_peptides'
    ]
    .fillna(0)
)


# ============================================================
# 6. SPECIES ORDER
# ============================================================

species_order = (
    plot_df
    .sort_values(
        'SL',
        ascending=True
    )['species']
    .tolist()
)


# ============================================================
# 7. FAMILY LOOKUP
# ============================================================

species_family_lookup = (
    plot_df
    .set_index('species')['family']
    .to_dict()
)


# ============================================================
# 8. CREATE TWO-PANEL FIGURE
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(16, 7),
    sharey=True,
    gridspec_kw={
        'width_ratios': [1, 1.15]
    }
)


# ============================================================
# PANEL A
# POPULATION-LEVEL SEROREACTIVITY
# ============================================================

ax = axes[0]

y = np.arange(
    len(plot_df)
)


# ------------------------------------------------------------
# HBDB
# ------------------------------------------------------------

if 'HBDB' in plot_df.columns:

    ax.scatter(
        plot_df['HBDB'],
        y,
        s=65,
        label='US healthy',
        zorder=3
    )


# ------------------------------------------------------------
# SL
# ------------------------------------------------------------

ax.scatter(
    plot_df['SL'],
    y,
    s=65,
    label='SL',
    zorder=3
)


# ------------------------------------------------------------
# Connecting lines
# ------------------------------------------------------------

if 'HBDB' in plot_df.columns:

    for i, (_, row) in enumerate(
        plot_df.iterrows()
    ):

        ax.plot(
            [
                row['HBDB'],
                row['SL']
            ],
            [
                i,
                i
            ],
            linewidth=1,
            alpha=0.45,
            zorder=1
        )


# ------------------------------------------------------------
# Y labels
# ------------------------------------------------------------

labels = [
    f"{species} ({species_family_lookup[species]})"
    for species in plot_df['species']
]

ax.set_yticks(y)

ax.set_yticklabels(
    labels,
    fontsize=11
)


# ------------------------------------------------------------
# Axis
# ------------------------------------------------------------

ax.xaxis.set_major_formatter(
    plt.FuncFormatter(
        lambda x, pos: f'{x:.0%}'
    )
)

ax.set_xlabel(
    'Fraction seroreactive',
    fontsize=13
)

ax.set_ylabel('')


ax.set_title(
    'Species-level seroreactivity',
    fontsize=15,
    pad=12
)


ax.legend(
    frameon=False,
    loc='lower right'
)


# ============================================================
# PANEL B
# NUMBER OF REACTIVE PEPTIDES
# ============================================================

ax = axes[1]


# Swarm plot
sns.swarmplot(
    data=peptide_breadth,
    x='n_reactive_peptides',
    y='species',
    order=species_order,
    size=4,
    alpha=0.65,
    ax=ax
)


# ------------------------------------------------------------
# Median number of reactive peptides
# ------------------------------------------------------------

medians = (
    peptide_breadth
    .groupby('species')[
        'n_reactive_peptides'
    ]
    .median()
    .reindex(species_order)
)


for i, median in enumerate(
    medians
):

    if pd.notna(median):

        ax.scatter(
            median,
            i,
            marker='D',
            s=65,
            zorder=10,
            edgecolor='black',
            linewidth=0.5
        )


# ------------------------------------------------------------
# Axis
# ------------------------------------------------------------

ax.set_xlabel(
    'Number of reactive peptides per seropositive SL individual',
    fontsize=13
)

ax.set_ylabel('')

ax.set_title(
    'Depth of peptide-level recognition',
    fontsize=15,
    pad=12
)


# Force integer x-axis
max_peptides = int(
    peptide_breadth[
        'n_reactive_peptides'
    ].max()
)

ax.set_xticks(
    np.arange(
        1,
        max_peptides + 1
    )
)


# ============================================================
# 9. FAMILY SEPARATORS
# ============================================================

# Species are already ordered according to plot_df,
# so identify family blocks.

family_sequence = [
    species_family_lookup[species]
    for species in species_order
]


family_positions = []

current_family = None
start_idx = 0

for i, family in enumerate(
    family_sequence
):

    if family != current_family:

        if current_family is not None:

            family_positions.append(
                (
                    current_family,
                    start_idx,
                    i - 1
                )
            )

        current_family = family
        start_idx = i


family_positions.append(
    (
        current_family,
        start_idx,
        len(family_sequence) - 1
    )
)


# ------------------------------------------------------------
# Draw separators in both panels
# ------------------------------------------------------------

for family, start, end in family_positions[:-1]:

    boundary = end + 0.5

    axes[0].axhline(
        boundary,
        linewidth=1,
        alpha=0.3
    )

    axes[1].axhline(
        boundary,
        linewidth=1,
        alpha=0.3
    )


# ============================================================
# 10. CLEAN UP
# ============================================================

for ax in axes:

    sns.despine(
        ax=ax
    )

    ax.tick_params(
        axis='y',
        length=0
    )


axes[0].tick_params(
    axis='x',
    labelsize=11
)

axes[1].tick_params(
    axis='x',
    labelsize=11
)


# ============================================================
# 11. OVERALL TITLE
# ============================================================

fig.suptitle(
    'Population prevalence and peptide-level breadth of viral seroreactivity',
    fontsize=18,
    y=1.02
)


plt.tight_layout()

plt.show()
```

### individual x species z-score heatmap

```python
# ============================================================
# 1. BUILD SPECIES x INDIVIDUAL MATRIX
# ============================================================

species_person_z_all = (
    z_long
    .groupby(['species', 'individual'])['z_score']
    .max()
    .unstack()
)

# Cohort assignment
individual_cohort = (
    z_long[['individual', 'cohort']]
    .drop_duplicates()
    .set_index('individual')['cohort']
)

# Keep individuals with known cohort
species_person_z_all = species_person_z_all.loc[
    :,
    species_person_z_all.columns.intersection(
        individual_cohort.index
    )
]

individual_cohort = individual_cohort.loc[
    species_person_z_all.columns
]

# Explicit cohort order
sl_individuals = (
    individual_cohort[
        individual_cohort == 'SL'
    ]
    .index
    .tolist()
)

hbdb_individuals = (
    individual_cohort[
        individual_cohort == 'HBDB'
    ]
    .index
    .tolist()
)

print(f"SL individuals: {len(sl_individuals)}")
print(f"HBDB individuals: {len(hbdb_individuals)}")


# ============================================================
# 2. CLUSTER INDIVIDUALS
# ============================================================

from scipy.cluster.hierarchy import linkage, leaves_list
from scipy.spatial.distance import pdist

cluster_matrix_all = (
    species_person_z_all
    .fillna(0)
)

# SL
sl_matrix = cluster_matrix_all[sl_individuals].T

sl_linkage = linkage(
    pdist(
        sl_matrix.values,
        metric='euclidean'
    ),
    method='ward'
)

sl_ordered_individuals = (
    sl_matrix.index[
        leaves_list(sl_linkage)
    ]
    .tolist()
)

# HBDB
hbdb_matrix = cluster_matrix_all[hbdb_individuals].T

hbdb_linkage = linkage(
    pdist(
        hbdb_matrix.values,
        metric='euclidean'
    ),
    method='ward'
)

hbdb_ordered_individuals = (
    hbdb_matrix.index[
        leaves_list(hbdb_linkage)
    ]
    .tolist()
)

ordered_individuals_all = (
    sl_ordered_individuals
    + hbdb_ordered_individuals
)


# ============================================================
# 3. CALCULATE SL SPECIES SEROPREVALENCE
# ============================================================

sl_species_prevalence = (
    species_individual[
        species_individual['cohort'] == 'SL'
    ]
    .groupby('species')['seroreactive']
    .mean()
    .rename('sl_prevalence')
    .reset_index()
)


# ============================================================
# 4. ADD TAXONOMY
# ============================================================

species_taxonomy = (
    hits[['species', 'genus', 'family']]
    .drop_duplicates('species')
)

species_order_df = (
    species_taxonomy
    .merge(
        sl_species_prevalence,
        on='species',
        how='left'
    )
)

species_order_df['sl_prevalence'] = (
    species_order_df['sl_prevalence']
    .fillna(0)
)


# ============================================================
# 5. TOP 2 SPECIES WITHIN EACH FAMILY
# ============================================================

TOP_N_PER_FAMILY = 2

top_species_by_family = (
    species_order_df
    .sort_values(
        ['family', 'sl_prevalence'],
        ascending=[True, False]
    )
    .groupby(
        'family',
        observed=True,
        group_keys=False
    )
    .head(TOP_N_PER_FAMILY)
    .copy()
)


# ============================================================
# 6. ORDER FAMILIES
#
# Family order is based on the strongest SL prevalence
# among the selected species in that family.
#
# This means families with the strongest representative
# species appear first.
# ============================================================

family_order = (
    top_species_by_family
    .groupby('family')['sl_prevalence']
    .max()
    .sort_values(ascending=False)
    .index
    .tolist()
)


# ============================================================
# 7. ORDER SPECIES WITHIN EACH FAMILY
# ============================================================

species_order_reduced = []

for family in family_order:

    family_species = (
        top_species_by_family[
            top_species_by_family['family'] == family
        ]
        .sort_values(
            'sl_prevalence',
            ascending=False
        )['species']
        .tolist()
    )

    species_order_reduced.extend(
        family_species
    )


# ============================================================
# 8. BUILD FINAL MATRIX
# ============================================================

plot_matrix_all = (
    cluster_matrix_all[
        ordered_individuals_all
    ]
)

plot_matrix = (
    plot_matrix_all
    .loc[
        plot_matrix_all.index.intersection(
            species_order_reduced
        )
    ]
    .reindex(species_order_reduced)
)


# ============================================================
# 9. PLOT
# ============================================================

HEATMAP_VMAX = 8

plot_matrix = (
    plot_matrix
    .clip(upper=HEATMAP_VMAX)
)

n_sl = len(sl_ordered_individuals)
n_hbdb = len(hbdb_ordered_individuals)

fig = plt.figure(
    figsize=(50, 35)
)

# Main heatmap
ax_heat = fig.add_axes([
    0.08,
    0.10,
    0.67,
    0.78
])

# Separate colorbar
ax_cbar = fig.add_axes([
    0.92,
    0.23,
    0.015,
    0.50
])


# ------------------------------------------------------------
# HEATMAP
# ------------------------------------------------------------

sns.heatmap(
    plot_matrix,
    ax=ax_heat,
    cmap='Blues',
    vmin=3,
    vmax=HEATMAP_VMAX,
    xticklabels=False,
    yticklabels=True,
    linewidths=0,
    cbar=True,
    cbar_ax=ax_cbar
)


# ============================================================
# 10. SPECIES LABELS
# ============================================================

row_info = (
    top_species_by_family
    .set_index('species')
    .loc[plot_matrix.index]
)

species_labels = [
    f"{species}"
    for species in plot_matrix.index
]

ax_heat.set_yticklabels(
    species_labels,
    rotation=0,
    ha='right',
    va='center'
)

ax_heat.tick_params(
    axis='y',
    labelsize=42,
    length=0,
    pad=12
)


# ============================================================
# 11. AXIS LABELS
# ============================================================

ax_heat.set_xlabel(
    'Individuals',
    fontsize=36,
    labelpad=20
)

ax_heat.set_ylabel(
    'Viral species',
    fontsize=42,
    labelpad=25
)

ax_heat.set_title(
    'Individual viral seroreactivity profiles',
    fontsize=42,
    pad=30
)


# ============================================================
# 12. COHORT BOUNDARY
# ============================================================

ax_heat.axvline(
    n_sl,
    linewidth=4,
    color='white',
    zorder=5
)


# ============================================================
# 13. COHORT LABELS
# ============================================================

ax_heat.text(
    n_sl / 2,
    -1.8,
    'SL',
    ha='center',
    va='center',
    fontsize=32,
    fontweight='bold'
)

ax_heat.text(
    n_sl + n_hbdb / 2,
    -1.8,
    'US healthy',
    ha='center',
    va='center',
    fontsize=32,
    fontweight='bold'
)


# ============================================================
# 14. FAMILY LABELS
# ============================================================

# Determine where each family starts/ends
row_families = (
    row_info['family']
)

family_positions = []

current_family = None
start_idx = 0

for i, family in enumerate(row_families):

    if family != current_family:

        if current_family is not None:
            family_positions.append(
                (
                    current_family,
                    start_idx,
                    i - 1
                )
            )

        current_family = family
        start_idx = i

# Final family
family_positions.append(
    (
        current_family,
        start_idx,
        len(row_families) - 1
    )
)


# Add labels to right of heatmap
for family, start, end in family_positions:

    center = (
        (start + end + 1) / 2
    )

    ax_heat.text(
        plot_matrix.shape[1] + 1.5,
        center,
        family,
        va='center',
        ha='left',
        fontsize=40,
        fontweight='bold',
        clip_on=False
    )


# ============================================================
# 15. FAMILY SEPARATORS
# ============================================================

for family, start, end in family_positions[:-1]:

    ax_heat.axhline(
        end + 1,
        color='white',
        linewidth=3,
        alpha=0.9
    )


# ============================================================
# 16. COLORBAR
# ============================================================

ax_cbar.set_ylabel(
    'Maximum peptide z-score',
    fontsize=34,
    labelpad=25
)

ax_cbar.tick_params(
    labelsize=28,
    length=8
)


# ============================================================
# 17. CLEAN UP
# ============================================================

sns.despine(
    ax=ax_heat,
    right=False,
    top=True
)

plt.show()
```

### heatmap addressing if the viral seroreactivity profiles that distinguish the exposed population from healthy controls also distinguish individuals who survived Lassa fever from household contacts (top 2 seroprevalent species per family)

```python
# ============================================================
# FIGURE: VIRAL SEROREACTIVITY BY LASSA FEVER OUTCOME
#
# Survivors vs household contacts vs US healthy
#
# Statistical test:
#   Fisher's exact test, survivors vs household contacts
#   Benjamini-Hochberg FDR correction across species
#
# Species shown:
#   Same top-2-per-family species used in the heatmap
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import fisher_exact
from statsmodels.stats.multitest import multipletests


# ============================================================
# 1. SPECIES TO INCLUDE
# ============================================================

selected_species = list(
    species_order_reduced
)

print(
    f"Number of selected species: {len(selected_species)}"
)


# ============================================================
# 2. GET Z-SCORES FOR SELECTED SPECIES
# ============================================================

outcome_z = (
    z_long[
        z_long['species'].isin(selected_species)
    ]
    .copy()
)


# ============================================================
# 3. ASSIGN CLINICAL / REFERENCE GROUP
#
# S-    = Lassa fever survivor
# C-    = household contact
# HBDB- = US healthy
# ============================================================

def assign_outcome(individual):

    individual = str(individual)

    if individual.startswith('S-'):
        return 'Lassa survivor'

    elif individual.startswith('C-'):
        return 'Household contact'

    elif individual.startswith('HBDB-'):
        return 'US healthy'

    return np.nan


outcome_z['outcome'] = (
    outcome_z['individual']
    .map(assign_outcome)
)


# ============================================================
# 4. SPECIES-LEVEL SEROREACTIVITY
#
# One person is seroreactive to a species if at least one
# retained peptide has z > 3.
# ============================================================

species_individual_outcome = (
    outcome_z
    .groupby(
        [
            'species',
            'individual',
            'outcome'
        ],
        observed=True
    )['z_score']
    .max()
    .reset_index()
)

species_individual_outcome['seroreactive'] = (
    species_individual_outcome['z_score'] > 3
)


# ============================================================
# 5. CALCULATE SPECIES-LEVEL PREVALENCE
# ============================================================

species_outcome_prevalence = (
    species_individual_outcome
    .groupby(
        [
            'species',
            'outcome'
        ],
        observed=True
    )
    .agg(
        n_seroreactive=(
            'seroreactive',
            'sum'
        ),
        n_individuals=(
            'seroreactive',
            'size'
        ),
        fraction_seroreactive=(
            'seroreactive',
            'mean'
        )
    )
    .reset_index()
)


# ============================================================
# 6. ADD TAXONOMY
# ============================================================

species_taxonomy_outcome = (
    hits[
        [
            'species',
            'genus',
            'family'
        ]
    ]
    .drop_duplicates('species')
)

species_outcome_prevalence = (
    species_outcome_prevalence
    .merge(
        species_taxonomy_outcome,
        on='species',
        how='left'
    )
)


# ============================================================
# 7. STATISTICAL TEST:
#    SURVIVOR vs HOUSEHOLD CONTACT
# ============================================================

results = []

for species in selected_species:

    sub = species_individual_outcome[
        species_individual_outcome['species'] == species
    ].copy()

    # ----------------------------
    # Counts
    # ----------------------------

    survivor_positive = (
        (
            (sub['outcome'] == 'Lassa survivor') &
            (sub['seroreactive'])
        )
        .sum()
    )

    survivor_negative = (
        (
            (sub['outcome'] == 'Lassa survivor') &
            (~sub['seroreactive'])
        )
        .sum()
    )

    contact_positive = (
        (
            (sub['outcome'] == 'Household contact') &
            (sub['seroreactive'])
        )
        .sum()
    )

    contact_negative = (
        (
            (sub['outcome'] == 'Household contact') &
            (~sub['seroreactive'])
        )
        .sum()
    )

    n_survivors = (
        survivor_positive
        + survivor_negative
    )

    n_contacts = (
        contact_positive
        + contact_negative
    )

    # ----------------------------
    # Fisher exact test
    # ----------------------------

    table = np.array([
        [
            survivor_positive,
            survivor_negative
        ],
        [
            contact_positive,
            contact_negative
        ]
    ])

    odds_ratio, p_value = fisher_exact(
        table,
        alternative='two-sided'
    )

    # ----------------------------
    # Fractions
    # ----------------------------

    survivor_fraction = (
        survivor_positive / n_survivors
        if n_survivors > 0
        else np.nan
    )

    contact_fraction = (
        contact_positive / n_contacts
        if n_contacts > 0
        else np.nan
    )

    results.append({
        'species': species,
        'survivor_positive': survivor_positive,
        'survivor_total': n_survivors,
        'survivor_fraction': survivor_fraction,
        'contact_positive': contact_positive,
        'contact_total': n_contacts,
        'contact_fraction': contact_fraction,
        'odds_ratio': odds_ratio,
        'p_value': p_value
    })


species_stats = pd.DataFrame(
    results
)


# ============================================================
# 8. FDR CORRECTION
# ============================================================

species_stats['q_value'] = (
    multipletests(
        species_stats['p_value'],
        method='fdr_bh'
    )[1]
)


# ============================================================
# 9. ADD TAXONOMY + DISPLAY ORDER
# ============================================================

species_stats = (
    species_stats
    .merge(
        species_taxonomy_outcome,
        on='species',
        how='left'
    )
)

species_stats['species'] = pd.Categorical(
    species_stats['species'],
    categories=selected_species,
    ordered=True
)

species_stats = (
    species_stats
    .sort_values('species')
    .reset_index(drop=True)
)


# ============================================================
# 10. SIGNIFICANCE LABEL
# ============================================================

species_stats['significance'] = np.select(
    [
        species_stats['q_value'] < 0.001,
        species_stats['q_value'] < 0.01,
        species_stats['q_value'] < 0.05
    ],
    [
        '***',
        '**',
        '*'
    ],
    default=''
)


# ============================================================
# 11. PREPARE PLOTTING TABLE
# ============================================================

plot_df = (
    species_stats[
        [
            'species',
            'family',
            'survivor_fraction',
            'contact_fraction',
            'odds_ratio',
            'p_value',
            'q_value',
            'significance'
        ]
    ]
    .copy()
)

# Plot from low to high prevalence so the highest signal
# appears at the top
plot_df = (
    plot_df
    .sort_values(
        'survivor_fraction',
        ascending=True
    )
    .reset_index(drop=True)
)


# ============================================================
# 12. LABELS
# ============================================================

plot_df['label'] = (
    plot_df['species'].astype(str)
    + '  ('
    + plot_df['family'].astype(str)
    + ')'
)

# Add significance to the label
plot_df['label'] = [
    f"{label}  {sig}"
    if sig != ''
    else label
    for label, sig
    in zip(
        plot_df['label'],
        plot_df['significance']
    )
]


# ============================================================
# 13. PLOT
# ============================================================

fig, ax = plt.subplots(
    figsize=(
        11,
        max(
            6,
            0.55 * len(plot_df)
        )
    )
)

y = np.arange(
    len(plot_df)
)


# ------------------------------------------------------------
# Connecting lines:
# household contact -> survivor
# ------------------------------------------------------------

for i, row in plot_df.iterrows():

    ax.plot(
        [
            row['contact_fraction'],
            row['survivor_fraction']
        ],
        [
            i,
            i
        ],
        linewidth=1.5,
        alpha=0.45,
        zorder=1
    )


# ------------------------------------------------------------
# Household contacts
# ------------------------------------------------------------

ax.scatter(
    plot_df['contact_fraction'],
    y,
    s=85,
    label='Household contact',
    zorder=3
)


# ------------------------------------------------------------
# Survivors
# ------------------------------------------------------------

ax.scatter(
    plot_df['survivor_fraction'],
    y,
    s=95,
    label='Lassa survivor',
    zorder=4
)


# ------------------------------------------------------------
# Species labels
# ------------------------------------------------------------

ax.set_yticks(y)

ax.set_yticklabels(
    plot_df['label'],
    fontsize=13
)

ax.set_xlabel(
    'Fraction seroreactive',
    fontsize=15
)

ax.set_ylabel('')

ax.xaxis.set_major_formatter(
    plt.FuncFormatter(
        lambda x, pos: f'{x:.0%}'
    )
)


# ============================================================
# 14. SIGNIFICANCE ANNOTATION
#
# Put q-value significance just beyond the larger point.
# ============================================================

for i, row in plot_df.iterrows():

    if row['significance'] == '':
        continue

    x_position = max(
        row['survivor_fraction'],
        row['contact_fraction']
    )

    ax.text(
        x_position + 0.012,
        i,
        row['significance'],
        ha='left',
        va='center',
        fontsize=14,
        fontweight='bold'
    )


# ============================================================
# 15. TITLE
# ============================================================

ax.set_title(
    (
        'Viral seroreactivity by Lassa fever outcome\n'
        'Top two species within each viral family'
    ),
    fontsize=17,
    pad=18
)


# ============================================================
# 16. LEGEND
# ============================================================

ax.legend(
    frameon=False,
    fontsize=11,
    loc='lower right'
)


# ============================================================
# 17. FOOTNOTE
# ============================================================

ax.text(
    0,
    -0.10,
    (
        'Fisher exact test: survivor vs household contact; '
        '* FDR < 0.05, ** FDR < 0.01, *** FDR < 0.001'
    ),
    transform=ax.transAxes,
    fontsize=10
)


# ============================================================
# 18. CLEAN UP
# ============================================================

sns.despine(
    ax=ax
)

ax.tick_params(
    axis='x',
    labelsize=11
)

plt.tight_layout()

plt.show()


# ============================================================
# 19. FULL STATISTICAL RESULTS TABLE
# ============================================================

display(
    species_stats[
        [
            'species',
            'family',
            'survivor_positive',
            'survivor_total',
            'survivor_fraction',
            'contact_positive',
            'contact_total',
            'contact_fraction',
            'odds_ratio',
            'p_value',
            'q_value',
            'significance'
        ]
    ]
    .sort_values(
        'q_value'
    )
)
```

### introducing clustering

```python
K = 8

def get_kmers(sequence, k=K):

    sequence = str(sequence).upper()

    return {
        sequence[i:i+k]
        for i in range(len(sequence) - k + 1)
    }


kmer_to_peptides = defaultdict(list)

for peptide, sequence in hits[
    ['peptide', 'sequence']
].itertuples(index=False):

    for kmer in get_kmers(sequence):

        kmer_to_peptides[kmer].append(
            peptide
        )
```

### construct graph

```python
G = nx.Graph()

# Include every filtered peptide,
# including singleton peptides
G.add_nodes_from(
    hits['peptide']
)

for peptides in kmer_to_peptides.values():

    if len(peptides) > 1:

        for p1, p2 in combinations(
            peptides,
            2
        ):

            G.add_edge(
                p1,
                p2
            )
```

### connect components

```python
components = list(
    nx.connected_components(G)
)

cluster_assignments = []

for cluster_id, peptides in enumerate(
    components
):

    for peptide in peptides:

        cluster_assignments.append({
            'peptide': peptide,
            'cluster_id': cluster_id
        })

cluster_assignments = pd.DataFrame(
    cluster_assignments
)
```

### attach clusters to the long z-score dataframe

```python
z_long_clustered = z_long.merge(
    cluster_assignments,
    on='peptide',
    how='left'
)
```

```python
cluster_taxonomy = (
    z_long_clustered
    [
        [
            'cluster_id',
            'peptide',
            'species',
            'genus',
            'family'
        ]
    ]
    .drop_duplicates()
    .groupby('cluster_id')
    .agg(
        n_peptides=('peptide', 'nunique'),
        n_species=('species', 'nunique'),
        n_genera=('genus', 'nunique'),
        n_families=('family', 'nunique')
    )
    .reset_index()
)
```

```python
# ============================================================
# SETUP FOR PEPTIDE SEQUENCE NETWORK ANALYSIS
#
# Creates:
#   network_hits
#   G_all
#   components_all
#
# Starting from ALL peptides retained in `hits`
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx

from collections import defaultdict
from itertools import combinations


# ============================================================
# 1. ALL FILTERED PEPTIDES
# ============================================================

network_hits = (
    hits[
        [
            'peptide',
            'sequence',
            'species',
            'genus',
            'family',
            'protein_name'
        ]
    ]
    .drop_duplicates('peptide')
    .copy()
)

print(
    f"Filtered peptides: "
    f"{network_hits['peptide'].nunique()}"
)

print(
    f"Species: "
    f"{network_hits['species'].nunique()}"
)

print(
    f"Genera: "
    f"{network_hits['genus'].nunique()}"
)

print(
    f"Families: "
    f"{network_hits['family'].nunique()}"
)


# ============================================================
# 2. CALCULATE SL PREVALENCE FOR EACH PEPTIDE
#
# fraction of SL individuals with z > 3 for that peptide
# ============================================================

sl_peptide_prevalence = (
    z_long[
        z_long['cohort'] == 'SL'
    ]
    .assign(
        reactive=lambda d:
        d['z_score'] > 3
    )
    .groupby('peptide')['reactive']
    .mean()
    .rename('sl_prevalence')
    .reset_index()
)


network_hits = (
    network_hits
    .merge(
        sl_peptide_prevalence,
        on='peptide',
        how='left'
    )
)

network_hits['sl_prevalence'] = (
    network_hits['sl_prevalence']
    .fillna(0)
)


# ============================================================
# 3. FUNCTION TO EXTRACT EXACT 8-AA K-MERS
# ============================================================

K = 8


def get_kmers(sequence, k=K):

    sequence = str(
        sequence
    ).upper()

    return {
        sequence[i:i+k]
        for i in range(
            len(sequence) - k + 1
        )
    }


# ============================================================
# 4. MAP EACH 8-MER TO THE PEPTIDES CONTAINING IT
# ============================================================

kmer_to_peptides = defaultdict(list)


for peptide, sequence in (
    network_hits[
        ['peptide', 'sequence']
    ]
    .itertuples(index=False)
):

    for kmer in get_kmers(
        sequence
    ):

        kmer_to_peptides[
            kmer
        ].append(
            peptide
        )


# ============================================================
# 5. BUILD PEPTIDE GRAPH
#
# Node = peptide
# Edge = two peptides share >=1 exact 8-aa k-mer
# ============================================================

G_all = nx.Graph()


# ------------------------------------------------------------
# Add nodes WITH metadata
# ------------------------------------------------------------

for _, row in network_hits.iterrows():

    G_all.add_node(
        row['peptide'],

        sequence=row['sequence'],

        species=row['species'],

        genus=row['genus'],

        family=row['family'],

        protein_name=row['protein_name'],

        sl_prevalence=row['sl_prevalence']
    )


# ------------------------------------------------------------
# Add edges
# ------------------------------------------------------------

for kmer, peptides in (
    kmer_to_peptides.items()
):

    # Need at least two peptides
    if len(peptides) < 2:
        continue

    # Avoid duplicate peptide IDs
    peptides = list(
        set(peptides)
    )

    for p1, p2 in combinations(
        peptides,
        2
    ):

        # ----------------------------------------
        # If edge already exists, record
        # additional shared 8-mer
        # ----------------------------------------

        if G_all.has_edge(
            p1,
            p2
        ):

            G_all[p1][p2][
                'shared_kmers'
            ].append(
                kmer
            )

        else:

            G_all.add_edge(
                p1,
                p2,
                shared_kmers=[
                    kmer
                ]
            )


# ============================================================
# 6. DEFINE CONNECTED COMPONENTS
#
# Each connected component = one sequence-linked
# peptide cluster
# ============================================================

components_all = list(
    nx.connected_components(
        G_all
    )
)


# ============================================================
# 7. ATTACH CLUSTER ID TO EACH PEPTIDE
# ============================================================

cluster_lookup = {}


for cluster_id, component in enumerate(
    components_all
):

    for peptide in component:

        cluster_lookup[
            peptide
        ] = cluster_id


network_hits[
    'cluster_id'
] = (
    network_hits[
        'peptide'
    ]
    .map(
        cluster_lookup
    )
)


# ============================================================
# 8. SUMMARY
# ============================================================

print()
print(
    f"Graph nodes: "
    f"{G_all.number_of_nodes()}"
)

print(
    f"Graph edges: "
    f"{G_all.number_of_edges()}"
)

print(
    f"Total sequence-linked clusters: "
    f"{len(components_all)}"
)

print(
    f"Singleton clusters: "
    f"{sum(len(x) == 1 for x in components_all)}"
)

print(
    f"Multi-peptide clusters: "
    f"{sum(len(x) > 1 for x in components_all)}"
)
```

```python
# ============================================================
# TWO REPRESENTATIVE SEQUENCE-LINKED PEPTIDE NETWORKS
#
# PANEL A:
#   Mammarenavirus cluster containing
#   Mammarenavirus lassaense
#
# PANEL B:
#   Orthonairovirus cluster containing
#   Orthonairovirus haemorrhagiae
#
# Nodes:
#   peptide
#
# Edges:
#   shared exact 8-aa k-mer
#
# Node color:
#   viral species
#
# Node shape:
#   protein
#
# Node size:
#   fraction of SL individuals reactive to peptide
#
# No peptide labels
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx

from matplotlib.lines import Line2D


# ============================================================
# 1. TARGET SPECIES
# ============================================================

TARGET_MAMMA = 'Mammarenavirus lassaense'
TARGET_ORTHO = 'Orthonairovirus haemorrhagiae'


# ============================================================
# 2. SUMMARIZE ALL EXISTING SEQUENCE-LINKED CLUSTERS
#
# Assumes you have already created:
#
#   components_all
#   network_hits
#   G_all
#
# from the all-filtered-peptide 8-mer analysis.
# ============================================================

cluster_records = []

for cluster_id, component in enumerate(
    components_all
):

    sub = (
        network_hits[
            network_hits['peptide'].isin(
                component
            )
        ]
        .copy()
    )

    species = sorted(
        sub['species']
        .dropna()
        .unique()
        .tolist()
    )

    genera = sorted(
        sub['genus']
        .dropna()
        .unique()
        .tolist()
    )

    families = sorted(
        sub['family']
        .dropna()
        .unique()
        .tolist()
    )

    proteins = sorted(
        sub['protein_name']
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    cluster_records.append({
        'cluster_id': cluster_id,
        'n_peptides': sub['peptide'].nunique(),
        'n_species': len(species),
        'n_genera': len(genera),
        'n_families': len(families),
        'species': species,
        'genera': genera,
        'families': families,
        'proteins': proteins
    })


cluster_summary = pd.DataFrame(
    cluster_records
)


# ============================================================
# 3. FUNCTION TO FIND TARGET-CONTAINING CLUSTER
# ============================================================

def select_target_cluster(
    target_species,
    target_genus,
    cluster_summary
):

    candidates = (
        cluster_summary[
            cluster_summary['species'].apply(
                lambda x:
                target_species in x
            )
        ]
        .copy()
    )

    # Keep only clusters confined to the target genus
    candidates = candidates[
        candidates['genera'].apply(
            lambda x:
            len(x) == 1
            and
            x[0] == target_genus
        )
    ]

    # Require at least 2 species so this is a
    # cross-species cluster
    candidates = candidates[
        candidates['n_species'] >= 2
    ]

    if len(candidates) == 0:
        return None

    # Prefer:
    # 1. greatest species breadth
    # 2. greatest peptide count

    candidates = (
        candidates
        .sort_values(
            [
                'n_species',
                'n_peptides'
            ],
            ascending=False
        )
        .reset_index(drop=True)
    )

    return candidates.iloc[0]


# ============================================================
# 4. SELECT THE TWO REPRESENTATIVE CLUSTERS
# ============================================================

mamma_cluster = select_target_cluster(
    TARGET_MAMMA,
    'Mammarenavirus',
    cluster_summary
)

ortho_cluster = select_target_cluster(
    TARGET_ORTHO,
    'Orthonairovirus',
    cluster_summary
)


# ============================================================
# 5. PRINT WHAT WAS SELECTED
# ============================================================

if mamma_cluster is None:

    print(
        f"No cross-species Mammarenavirus cluster "
        f"contains {TARGET_MAMMA}"
    )

else:

    print(
        "\nMAMMARENAVIRUS CLUSTER"
    )

    print(
        f"Cluster ID: {mamma_cluster['cluster_id']}"
    )

    print(
        f"Peptides: {mamma_cluster['n_peptides']}"
    )

    print(
        f"Species: {mamma_cluster['n_species']}"
    )

    print(
        f"Species represented: "
        f"{', '.join(mamma_cluster['species'])}"
    )

    print(
        f"Proteins: "
        f"{', '.join(mamma_cluster['proteins'])}"
    )


if ortho_cluster is None:

    print(
        f"No cross-species Orthonairovirus cluster "
        f"contains {TARGET_ORTHO}"
    )

else:

    print(
        "\nORTHONAIROVIRUS CLUSTER"
    )

    print(
        f"Cluster ID: {ortho_cluster['cluster_id']}"
    )

    print(
        f"Peptides: {ortho_cluster['n_peptides']}"
    )

    print(
        f"Species: {ortho_cluster['n_species']}"
    )

    print(
        f"Species represented: "
        f"{', '.join(ortho_cluster['species'])}"
    )

    print(
        f"Proteins: "
        f"{', '.join(ortho_cluster['proteins'])}"
    )


# ============================================================
# 6. PROTEIN CLASSIFICATION
# ============================================================

def protein_class(
    protein
):

    protein = str(
        protein
    ).lower()

    if (
        'nucleoprotein' in protein
        or protein == 'np'
    ):
        return 'Nucleoprotein'

    elif 'glycoprotein' in protein:
        return 'Glycoprotein'

    elif 'polymerase' in protein:
        return 'Polymerase'

    elif 'matrix' in protein:
        return 'Matrix'

    elif 'capsid' in protein:
        return 'Capsid'

    else:
        return 'Other'


protein_markers = {
    'Nucleoprotein': 'o',
    'Glycoprotein': 's',
    'Polymerase': '^',
    'Matrix': 'D',
    'Capsid': 'P',
    'Other': 'X'
}


# ============================================================
# 7. FUNCTION TO PLOT ONE CLUSTER
# ============================================================

def plot_target_cluster(
    ax,
    cluster_row,
    target_species,
    title
):

    cluster_id = int(
        cluster_row['cluster_id']
    )

    component = (
        components_all[
            cluster_id
        ]
    )

    G = (
        G_all
        .subgraph(component)
        .copy()
    )


    # --------------------------------------------------------
    # Layout
    # --------------------------------------------------------

    pos = nx.spring_layout(
        G,
        seed=42,
        k=1.2,
        iterations=500
    )


    # --------------------------------------------------------
    # Species
    # --------------------------------------------------------

    species_here = sorted({
        G.nodes[node]['species']
        for node in G.nodes()
    })


    species_palette = {
        species:
        plt.cm.tab20(
            i % 20
        )

        for i, species
        in enumerate(species_here)
    }


    # --------------------------------------------------------
    # Edges
    # --------------------------------------------------------

    nx.draw_networkx_edges(
        G,
        pos,
        ax=ax,
        width=1.1,
        alpha=0.35
    )


    # --------------------------------------------------------
    # Nodes by protein
    # --------------------------------------------------------

    for protein_type, marker in (
        protein_markers.items()
    ):

        nodes = [
            node

            for node in G.nodes()

            if protein_class(
                G.nodes[node]['protein_name']
            ) == protein_type
        ]

        if not nodes:
            continue


        colors = [
            species_palette[
                G.nodes[node]['species']
            ]

            for node in nodes
        ]


        sizes = [
            (
                100
                + 2200
                * G.nodes[node][
                    'sl_prevalence'
                ]
            )

            for node in nodes
        ]


        nx.draw_networkx_nodes(
            G,
            pos,
            ax=ax,
            nodelist=nodes,
            node_size=sizes,
            node_color=colors,
            node_shape=marker,
            edgecolors='black',
            linewidths=0.6,
            alpha=0.9
        )


    # --------------------------------------------------------
    # Highlight TARGET SPECIES
    #
    # Keep the same protein shape, but use a thicker outline.
    # --------------------------------------------------------

    for protein_type, marker in (
        protein_markers.items()
    ):

        target_nodes = [

            node

            for node in G.nodes()

            if (
                G.nodes[node]['species']
                == target_species
            )
            and
            protein_class(
                G.nodes[node][
                    'protein_name'
                ]
            ) == protein_type
        ]

        if not target_nodes:
            continue


        nx.draw_networkx_nodes(
            G,
            pos,
            ax=ax,
            nodelist=target_nodes,
            node_size=[
                (
                    130
                    + 2200
                    * G.nodes[node][
                        'sl_prevalence'
                    ]
                )

                for node in target_nodes
            ],
            node_color=[
                species_palette[
                    G.nodes[node]['species']
                ]

                for node in target_nodes
            ],
            node_shape=marker,
            edgecolors='black',
            linewidths=2.5,
            alpha=1
        )


    # --------------------------------------------------------
    # Metadata
    # --------------------------------------------------------

    genera_here = sorted({
        G.nodes[node]['genus']
        for node in G.nodes()
    })

    families_here = sorted({
        G.nodes[node]['family']
        for node in G.nodes()
    })

    proteins_here = sorted({
        protein_class(
            G.nodes[node]['protein_name']
        )

        for node in G.nodes()
    })


    metadata = (
        f"{G.number_of_nodes()} peptides\n"
        f"{len(species_here)} species\n"
        f"Genus: {', '.join(genera_here)}\n"
        f"Family: {', '.join(families_here)}\n"
        f"Proteins: {', '.join(proteins_here)}"
    )


    ax.text(
        0.02,
        0.02,
        metadata,
        transform=ax.transAxes,
        ha='left',
        va='bottom',
        fontsize=10,
        bbox=dict(
            boxstyle='round,pad=0.4',
            facecolor='white',
            alpha=0.9,
            edgecolor='none'
        )
    )


    ax.set_title(
        title,
        fontsize=16,
        pad=15
    )

    ax.axis('off')


    return (
        G,
        species_palette
    )


# ============================================================
# 8. CREATE TWO-PANEL FIGURE
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(18, 9)
)


# ============================================================
# PANEL A — MAMMARENAVIRUS / LASSA
# ============================================================

if mamma_cluster is not None:

    G_mamma, palette_mamma = (
        plot_target_cluster(
            axes[0],
            mamma_cluster,
            TARGET_MAMMA,
            'A  Mammarenavirus cluster containing Lassa'
        )
    )

else:

    axes[0].text(
        0.5,
        0.5,
        'No qualifying Mammarenavirus cluster',
        ha='center',
        va='center'
    )

    axes[0].axis('off')


# ============================================================
# PANEL B — ORTHONAIROVIRUS / HAEMORRHAGIAE
# ============================================================

if ortho_cluster is not None:

    G_ortho, palette_ortho = (
        plot_target_cluster(
            axes[1],
            ortho_cluster,
            TARGET_ORTHO,
            'B  Orthonairovirus cluster containing haemorrhagiae'
        )
    )

else:

    axes[1].text(
        0.5,
        0.5,
        'No qualifying Orthonairovirus cluster',
        ha='center',
        va='center'
    )

    axes[1].axis('off')


# ============================================================
# 9. LEGENDS
# ============================================================

# ---- species legends ----

if mamma_cluster is not None:

    handles_mamma = [
        Line2D(
            [0],
            [0],
            marker='o',
            linestyle='None',
            markerfacecolor=palette_mamma[species],
            markeredgecolor='black',
            markersize=8,
            label=species
        )

        for species
        in sorted(
            palette_mamma.keys()
        )
    ]

    axes[0].legend(
        handles=handles_mamma,
        title='Species',
        frameon=False,
        fontsize=8,
        title_fontsize=10,
        loc='upper left',
        bbox_to_anchor=(0, 1)
    )


if ortho_cluster is not None:

    handles_ortho = [
        Line2D(
            [0],
            [0],
            marker='o',
            linestyle='None',
            markerfacecolor=palette_ortho[species],
            markeredgecolor='black',
            markersize=8,
            label=species
        )

        for species
        in sorted(
            palette_ortho.keys()
        )
    ]

    axes[1].legend(
        handles=handles_ortho,
        title='Species',
        frameon=False,
        fontsize=8,
        title_fontsize=10,
        loc='upper left',
        bbox_to_anchor=(0, 1)
    )


# ============================================================
# PROTEIN LEGEND
# ============================================================

protein_handles = [
    Line2D(
        [0],
        [0],
        marker=marker,
        linestyle='None',
        markerfacecolor='white',
        markeredgecolor='black',
        markersize=9,
        label=protein_type
    )

    for protein_type, marker
    in protein_markers.items()
]


fig.legend(
    handles=protein_handles,
    title='Protein',
    frameon=False,
    fontsize=9,
    title_fontsize=10,
    loc='lower center',
    ncol=6,
    bbox_to_anchor=(0.5, -0.01)
)


# ============================================================
# OVERALL TITLE
# ============================================================

fig.suptitle(
    (
        'Representative sequence-linked peptide clusters '
        'containing Mammarenavirus lassaense and Orthonairovirus haemorrhagiae species'
    ),
    fontsize=19,
    y=1.02
)


plt.tight_layout()

plt.show()
```

```python
# ============================================================
# ALL FILTERED PEPTIDES: SEQUENCE-LINKAGE NETWORK
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx

from collections import defaultdict
from itertools import combinations
from matplotlib.lines import Line2D


# ============================================================
# 1. PEPTIDE DATA
# ============================================================

network_hits = (
    hits[
        [
            'peptide',
            'sequence',
            'species',
            'genus',
            'family',
            'protein_name'
        ]
    ]
    .drop_duplicates('peptide')
    .copy()
)

print(
    f"Filtered peptides: {network_hits['peptide'].nunique()}"
)

print(
    f"Species represented: {network_hits['species'].nunique()}"
)

print(
    f"Families represented: {network_hits['family'].nunique()}"
)


# ============================================================
# 2. PEPTIDE-LEVEL SL SEROREACTIVITY
# ============================================================

peptide_sl_prevalence = (
    z_long[
        z_long['cohort'] == 'SL'
    ]
    .groupby('peptide')['reactive']
    .mean()
    .rename('sl_prevalence')
    .reset_index()
)

network_hits = (
    network_hits
    .merge(
        peptide_sl_prevalence,
        on='peptide',
        how='left'
    )
)

network_hits['sl_prevalence'] = (
    network_hits['sl_prevalence']
    .fillna(0)
)


# ============================================================
# 3. EXACT 8-AA K-MERS
# ============================================================

K = 8


def get_kmers(sequence, k=K):

    sequence = str(sequence).upper()

    return {
        sequence[i:i+k]
        for i in range(len(sequence) - k + 1)
    }


kmer_to_peptides = defaultdict(list)

for peptide, sequence in (
    network_hits[
        ['peptide', 'sequence']
    ]
    .itertuples(index=False)
):

    for kmer in get_kmers(sequence):

        kmer_to_peptides[kmer].append(
            peptide
        )


# ============================================================
# 4. BUILD GRAPH
# ============================================================

G_all = nx.Graph()


# ------------------------------------------------------------
# Add nodes WITH metadata
# ------------------------------------------------------------

for _, row in network_hits.iterrows():

    G_all.add_node(
        row['peptide'],
        species=row['species'],
        genus=row['genus'],
        family=row['family'],
        protein_name=row['protein_name'],
        sl_prevalence=row['sl_prevalence']
    )


# ------------------------------------------------------------
# Add edges
#
# Peptides sharing an exact 8-aa k-mer are connected
# ------------------------------------------------------------

for peptides in kmer_to_peptides.values():

    if len(peptides) > 1:

        for peptide_a, peptide_b in combinations(
            peptides,
            2
        ):

            G_all.add_edge(
                peptide_a,
                peptide_b
            )


print(
    f"Nodes: {G_all.number_of_nodes()}"
)

print(
    f"Edges: {G_all.number_of_edges()}"
)


# ============================================================
# 5. CONNECTED COMPONENTS
# ============================================================

components_all = list(
    nx.connected_components(G_all)
)

print(
    f"Sequence-linked clusters: {len(components_all)}"
)


# Attach cluster IDs
cluster_lookup = {}

for cluster_id, component in enumerate(
    components_all
):

    for peptide in component:

        cluster_lookup[peptide] = cluster_id


network_hits['cluster_id'] = (
    network_hits['peptide']
    .map(cluster_lookup)
)


# ============================================================
# 6. SPECIES COLORS
# ============================================================

species_list = (
    network_hits['species']
    .dropna()
    .unique()
    .tolist()
)

species_to_color = {
    species: plt.cm.tab20(
        i % 20
    )
    for i, species in enumerate(
        species_list
    )
}


# Get species for every graph node
node_species = nx.get_node_attributes(
    G_all,
    'species'
)

node_colors = [
    species_to_color[
        node_species[node]
    ]
    for node in G_all.nodes()
]


# ============================================================
# 7. NODE SIZE
#
# Larger = recognized by a larger fraction of SL participants
# ============================================================

node_prevalence = nx.get_node_attributes(
    G_all,
    'sl_prevalence'
)

node_sizes = [
    20
    + 1200 * node_prevalence[node]
    for node in G_all.nodes()
]


# ============================================================
# 8. LAYOUT
# ============================================================

pos_all = nx.spring_layout(
    G_all,
    seed=42,
    k=0.25,
    iterations=300
)


# ============================================================
# 9. PLOT
# ============================================================

fig, ax = plt.subplots(
    figsize=(22, 20)
)


# ------------------------------------------------------------
# Edges
# ------------------------------------------------------------

nx.draw_networkx_edges(
    G_all,
    pos_all,
    ax=ax,
    width=0.5,
    alpha=0.18
)


# ------------------------------------------------------------
# Nodes
# ------------------------------------------------------------

nx.draw_networkx_nodes(
    G_all,
    pos_all,
    ax=ax,
    node_size=node_sizes,
    node_color=node_colors,
    edgecolors='black',
    linewidths=0.25,
    alpha=0.85
)


# ============================================================
# 10. LABEL ONLY HIGH-PREVALENCE PEPTIDES
#
# Labelling all 1,274 peptides will be unreadable.
# ============================================================

LABEL_PREVALENCE = 0.10

labels = {}

for node in G_all.nodes():

    prevalence = node_prevalence[node]

    if prevalence >= LABEL_PREVALENCE:

        labels[node] = node.split('|')[0]


nx.draw_networkx_labels(
    G_all,
    pos_all,
    labels=labels,
    ax=ax,
    font_size=5,
    alpha=0.8
)


# ============================================================
# 11. SPECIES LEGEND
# ============================================================

legend_handles = []

for species in species_list:

    legend_handles.append(
        Line2D(
            [0],
            [0],
            marker='o',
            color='none',
            markerfacecolor=species_to_color[species],
            markeredgecolor='black',
            markersize=7,
            label=species
        )
    )


ax.legend(
    handles=legend_handles,
    title='Viral species',
    frameon=False,
    fontsize=7,
    title_fontsize=10,
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)


# ============================================================
# 12. TITLE
# ============================================================

ax.set_title(
    'Sequence-linkage network of all high-specificity viral peptides',
    fontsize=18,
    pad=20
)


# ============================================================
# 13. CLEAN UP
# ============================================================

ax.axis('off')

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# CROSS-SPECIES SEQUENCE-LINKED PEPTIDE NETWORK
#
# Start from ALL filtered peptides,
# but visualize only connected components containing
# peptides from >=2 viral species.
#
# Node = peptide
# Edge = shared exact 8-aa k-mer
# Color = species
# Size = SL peptide prevalence
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from matplotlib.lines import Line2D


# ============================================================
# 1. FIND CROSS-SPECIES COMPONENTS
# ============================================================

cross_species_components = []

for component in nx.connected_components(G_all):

    species_in_component = {
        G_all.nodes[node]['species']
        for node in component
    }

    if len(species_in_component) >= 2:

        cross_species_components.append(
            component
        )


print(
    f"Cross-species components: "
    f"{len(cross_species_components)}"
)


# ============================================================
# 2. SORT COMPONENTS
#
# Largest / most taxonomically broad first
# ============================================================

cross_species_components = sorted(
    cross_species_components,
    key=lambda component: (
        len({
            G_all.nodes[node]['species']
            for node in component
        }),
        len(component)
    ),
    reverse=True
)


# ============================================================
# 3. OPTIONALLY KEEP TOP N COMPONENTS
#
# Set to None if you really want every cross-species component.
# For a paper figure I would start around 15-25.
# ============================================================

TOP_N_COMPONENTS = 20

if TOP_N_COMPONENTS is not None:

    cross_species_components = (
        cross_species_components[
            :TOP_N_COMPONENTS
        ]
    )


# ============================================================
# 4. BUILD SUBGRAPH
# ============================================================

nodes_to_plot = set().union(
    *cross_species_components
)

G_cross = (
    G_all
    .subgraph(nodes_to_plot)
    .copy()
)


print(
    f"Peptides shown: {G_cross.number_of_nodes()}"
)

print(
    f"Edges shown: {G_cross.number_of_edges()}"
)


# ============================================================
# 5. CREATE A SEPARATE LAYOUT FOR EACH COMPONENT
#
# This is the important part.
#
# Rather than spring-layouting disconnected components
# together, arrange each cluster separately on a grid.
# ============================================================

n_components = len(
    cross_species_components
)

n_cols = 5

n_rows = int(
    np.ceil(
        n_components / n_cols
    )
)


fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(
        5 * n_cols,
        5 * n_rows
    )
)

axes = np.array(
    axes
).reshape(-1)


# ============================================================
# 6. SPECIES COLORS
# ============================================================

species_in_plot = sorted({
    G_cross.nodes[node]['species']
    for node in G_cross.nodes()
})

species_to_color_cross = {
    species: plt.cm.tab20(
        i % 20
    )
    for i, species in enumerate(
        species_in_plot
    )
}


# ============================================================
# 7. PLOT EACH COMPONENT
# ============================================================

for component_number, (
    ax,
    component
) in enumerate(
    zip(
        axes,
        cross_species_components
    ),
    start=1
):

    subgraph = (
        G_all
        .subgraph(component)
        .copy()
    )

    # ----------------------------------------
    # Layout within this component only
    # ----------------------------------------

    pos = nx.spring_layout(
        subgraph,
        seed=42,
        k=1.2,
        iterations=300
    )


    # ----------------------------------------
    # Node colors
    # ----------------------------------------

    node_colors = [
        species_to_color_cross[
            subgraph.nodes[node]['species']
        ]
        for node in subgraph.nodes()
    ]


    # ----------------------------------------
    # Node sizes
    # ----------------------------------------

    node_sizes = [
        100
        + 1000
        * subgraph.nodes[node][
            'sl_prevalence'
        ]
        for node in subgraph.nodes()
    ]


    # ----------------------------------------
    # Edges
    # ----------------------------------------

    nx.draw_networkx_edges(
        subgraph,
        pos,
        ax=ax,
        width=1,
        alpha=0.35
    )


    # ----------------------------------------
    # Nodes
    # ----------------------------------------

    nx.draw_networkx_nodes(
        subgraph,
        pos,
        ax=ax,
        node_size=node_sizes,
        node_color=node_colors,
        edgecolors='black',
        linewidths=0.5,
        alpha=0.9
    )


    # ----------------------------------------
    # Cluster information
    # ----------------------------------------

    species_here = {
        subgraph.nodes[node]['species']
        for node in subgraph.nodes()
    }

    genera_here = {
        subgraph.nodes[node]['genus']
        for node in subgraph.nodes()
    }

    families_here = {
        subgraph.nodes[node]['family']
        for node in subgraph.nodes()
    }


    ax.set_title(
        (
            f'Cluster {component_number}\n'
            f'{subgraph.number_of_nodes()} peptides | '
            f'{len(species_here)} species | '
            f'{len(genera_here)} genera'
        ),
        fontsize=11
    )

    ax.axis('off')


# ============================================================
# 8. REMOVE UNUSED PANELS
# ============================================================

for ax in axes[
    len(cross_species_components):
]:

    ax.axis('off')


# ============================================================
# 9. OVERALL TITLE
# ============================================================

fig.suptitle(
    (
        'Cross-species sequence-linked peptide networks '
        'among high-specificity viral hits'
    ),
    fontsize=18,
    y=1.01
)


plt.tight_layout()

plt.show()
```

```python
# ============================================================
# ENHANCED CROSS-SPECIES NETWORK FIGURE
#
# Node color  = species
# Node shape  = protein
# Node size   = SL peptide seroreactivity
# ============================================================

from matplotlib.lines import Line2D

# ------------------------------------------------------------
# Protein categories
# ------------------------------------------------------------

def protein_class(protein):

    protein = str(protein).lower()

    if 'nucleoprotein' in protein or protein == 'np':
        return 'Nucleoprotein'

    elif 'glycoprotein' in protein or 'glycoprotein precursor' in protein:
        return 'Glycoprotein'

    elif 'polymerase' in protein:
        return 'Polymerase'

    elif 'matrix' in protein:
        return 'Matrix'

    elif 'capsid' in protein:
        return 'Capsid'

    else:
        return 'Other'


protein_markers = {
    'Nucleoprotein': 'o',
    'Glycoprotein': 's',
    'Polymerase': '^',
    'Matrix': 'D',
    'Capsid': 'P',
    'Other': 'X'
}


# ------------------------------------------------------------
# Species colors
# ------------------------------------------------------------

species_in_plot = sorted({
    G_all.nodes[node]['species']
    for component in cross_species_components
    for node in component
})

species_to_color_cross = {
    species: plt.cm.tab20(
        i % 20
    )
    for i, species in enumerate(
        species_in_plot
    )
}


# ------------------------------------------------------------
# Figure layout
# ------------------------------------------------------------

n_components = len(
    cross_species_components
)

n_cols = 4

n_rows = int(
    np.ceil(
        n_components / n_cols
    )
)

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(
        6 * n_cols,
        6 * n_rows
    )
)

axes = np.array(
    axes
).reshape(-1)


# ============================================================
# PLOT EACH CLUSTER
# ============================================================

for cluster_number, (
    ax,
    component
) in enumerate(
    zip(
        axes,
        cross_species_components
    ),
    start=1
):

    subgraph = (
        G_all
        .subgraph(component)
        .copy()
    )

    # --------------------------------------------------------
    # Layout
    # --------------------------------------------------------

    pos = nx.spring_layout(
        subgraph,
        seed=42,
        k=1.2,
        iterations=300
    )


    # --------------------------------------------------------
    # Edges
    # --------------------------------------------------------

    nx.draw_networkx_edges(
        subgraph,
        pos,
        ax=ax,
        width=1.2,
        alpha=0.35
    )


    # --------------------------------------------------------
    # Draw nodes by protein class
    # --------------------------------------------------------

    for protein_type, marker in (
        protein_markers.items()
    ):

        nodes = [
            node
            for node in subgraph.nodes()
            if protein_class(
                subgraph.nodes[node][
                    'protein_name'
                ]
            ) == protein_type
        ]

        if len(nodes) == 0:
            continue

        sizes = [
            (
                100
                + 1200
                * subgraph.nodes[node][
                    'sl_prevalence'
                ]
            )
            for node in nodes
        ]

        colors = [
            species_to_color_cross[
                subgraph.nodes[node]['species']
            ]
            for node in nodes
        ]

        nx.draw_networkx_nodes(
            subgraph,
            pos,
            nodelist=nodes,
            ax=ax,
            node_size=sizes,
            node_color=colors,
            node_shape=marker,
            edgecolors='black',
            linewidths=0.6,
            alpha=0.9
        )


    # ========================================================
    # CLUSTER METADATA
    # ========================================================

    species_here = sorted({
        subgraph.nodes[node]['species']
        for node in subgraph.nodes()
    })

    genera_here = sorted({
        subgraph.nodes[node]['genus']
        for node in subgraph.nodes()
    })

    families_here = sorted({
        subgraph.nodes[node]['family']
        for node in subgraph.nodes()
    })

    proteins_here = sorted({
        protein_class(
            subgraph.nodes[node]['protein_name']
        )
        for node in subgraph.nodes()
    })


    # --------------------------------------------------------
    # Build readable metadata string
    # --------------------------------------------------------

    if len(genera_here) <= 3:

        genus_text = ', '.join(
            genera_here
        )

    else:

        genus_text = (
            f'{len(genera_here)} genera'
        )


    if len(proteins_here) <= 3:

        protein_text = ', '.join(
            proteins_here
        )

    else:

        protein_text = (
            f'{len(proteins_here)} protein classes'
        )


    # --------------------------------------------------------
    # Title
    # --------------------------------------------------------

    ax.set_title(
        (
            f'Cluster {cluster_number}\n'
            f'{len(species_here)} species | '
            f'{genus_text}\n'
            f'{subgraph.number_of_nodes()} peptides | '
            f'{protein_text}'
        ),
        fontsize=10,
        pad=10
    )


    # --------------------------------------------------------
    # Species labels
    #
    # Label only if the cluster is reasonably small
    # --------------------------------------------------------

    if subgraph.number_of_nodes() <= 20:

        labels = {}

        for node in subgraph.nodes():

            labels[node] = (
                subgraph.nodes[node][
                    'species'
                ]
            )

        nx.draw_networkx_labels(
            subgraph,
            pos,
            labels=labels,
            ax=ax,
            font_size=7
        )


    ax.axis('off')


# ============================================================
# REMOVE UNUSED AXES
# ============================================================

for ax in axes[
    len(cross_species_components):
]:

    ax.axis('off')


# ============================================================
# SPECIES LEGEND
# ============================================================

species_handles = [
    Line2D(
        [0],
        [0],
        marker='o',
        color='none',
        markerfacecolor=species_to_color_cross[species],
        markeredgecolor='black',
        markersize=7,
        label=species
    )
    for species in species_in_plot
]


# ============================================================
# PROTEIN LEGEND
# ============================================================

protein_handles = [
    Line2D(
        [0],
        [0],
        marker=marker,
        color='black',
        markerfacecolor='white',
        markersize=8,
        linestyle='None',
        label=protein_type
    )
    for protein_type, marker
    in protein_markers.items()
]


fig.legend(
    handles=(
        species_handles
        + protein_handles
    ),
    title='Species / protein',
    frameon=False,
    fontsize=7,
    title_fontsize=9,
    bbox_to_anchor=(1.01, 1),
    loc='upper left'
)


# ============================================================
# OVERALL TITLE
# ============================================================

fig.suptitle(
    (
        'Cross-species sequence-linked peptide networks '
        'among high-specificity viral hits'
    ),
    fontsize=18,
    y=1.01
)


plt.tight_layout()

plt.show()
```

```python
# ============================================================
# SEQUENCE CLUSTERING RESTRICTED TO SPECIES SHOWN IN HEATMAP
#
# Uses the exact species displayed in:
#     species_order_reduced
#
# Rebuilds the 8-mer sequence graph using only peptides
# assigned to those species.
# ============================================================

from collections import defaultdict
from itertools import combinations
import networkx as nx
import pandas as pd
import numpy as np


# ============================================================
# 1. DEFINE THE SPECIES SET
# ============================================================

heatmap_species = set(
    species_order_reduced
)

print(
    f"Number of heatmap species: {len(heatmap_species)}"
)


# ============================================================
# 2. RESTRICT FILTERED PEPTIDES TO THOSE SPECIES
# ============================================================

hits_heatmap = (
    hits[
        hits['species'].isin(
            heatmap_species
        )
    ]
    .copy()
)

print(
    f"Filtered peptides from heatmap species: "
    f"{hits_heatmap['peptide'].nunique()}"
)

print(
    f"Species represented: "
    f"{hits_heatmap['species'].nunique()}"
)


# ============================================================
# 3. EXTRACT EXACT 8-AA K-MERS
# ============================================================

K = 8

def get_kmers(
    sequence,
    k=K
):

    sequence = str(
        sequence
    ).upper()

    return {
        sequence[i:i+k]
        for i in range(
            len(sequence) - k + 1
        )
    }


kmer_to_peptides = defaultdict(list)

for peptide, sequence in (
    hits_heatmap[
        ['peptide', 'sequence']
    ]
    .itertuples(index=False)
):

    for kmer in get_kmers(
        sequence
    ):

        kmer_to_peptides[kmer].append(
            peptide
        )


# ============================================================
# 4. BUILD SEQUENCE-SIMILARITY GRAPH
# ============================================================

G_heatmap = nx.Graph()

# Include every peptide, including singletons
G_heatmap.add_nodes_from(
    hits_heatmap['peptide']
)

# Connect peptides sharing an exact 8-aa k-mer
for peptides in kmer_to_peptides.values():

    if len(peptides) > 1:

        for p1, p2 in combinations(
            peptides,
            2
        ):

            G_heatmap.add_edge(
                p1,
                p2
            )


# ============================================================
# 5. DEFINE CONNECTED COMPONENTS AS CLUSTERS
# ============================================================

components_heatmap = list(
    nx.connected_components(
        G_heatmap
    )
)

cluster_assignments_heatmap = []

for cluster_id, peptides in enumerate(
    components_heatmap
):

    for peptide in peptides:

        cluster_assignments_heatmap.append({
            'peptide': peptide,
            'cluster_id': cluster_id
        })


cluster_assignments_heatmap = pd.DataFrame(
    cluster_assignments_heatmap
)


# ============================================================
# 6. ATTACH CLUSTERS TO PEPTIDE METADATA
# ============================================================

hits_heatmap_z = (
    hits_heatmap
    .merge(
        cluster_assignments_heatmap,
        on='peptide',
        how='left'
    )
)


# ============================================================
# 7. CHARACTERIZE EACH CLUSTER
# ============================================================

cluster_taxonomy_heatmap = (
    hits_heatmap_z[
        [
            'cluster_id',
            'peptide',
            'species',
            'genus',
            'family'
        ]
    ]
    .drop_duplicates()
    .groupby('cluster_id')
    .agg(
        n_peptides=(
            'peptide',
            'nunique'
        ),
        n_species=(
            'species',
            'nunique'
        ),
        n_genera=(
            'genus',
            'nunique'
        ),
        n_families=(
            'family',
            'nunique'
        )
    )
    .reset_index()
)


# ============================================================
# 8. CLASSIFY CLUSTERS
# ============================================================

def classify_cluster(
    row
):

    if row['n_peptides'] == 1:

        return 'Singleton'

    elif row['n_species'] == 1:

        return 'Within species'

    elif row['n_genera'] == 1:

        return 'Cross-species\nwithin genus'

    elif row['n_families'] == 1:

        return 'Cross-genus\nwithin family'

    else:

        return 'Cross-family'


cluster_taxonomy_heatmap[
    'cluster_type'
] = (
    cluster_taxonomy_heatmap
    .apply(
        classify_cluster,
        axis=1
    )
)


# ============================================================
# 9. SUMMARY OF CLUSTER COMPOSITION
# ============================================================

cluster_order = [
    'Singleton',
    'Within species',
    'Cross-species\nwithin genus',
    'Cross-genus\nwithin family',
    'Cross-family'
]

cluster_counts_heatmap = (
    cluster_taxonomy_heatmap[
        'cluster_type'
    ]
    .value_counts()
    .reindex(
        cluster_order
    )
    .fillna(0)
)

cluster_percent_heatmap = (
    100
    * cluster_counts_heatmap
    / cluster_counts_heatmap.sum()
)

cluster_summary_heatmap = pd.DataFrame({
    'n_clusters': cluster_counts_heatmap,
    'percent': cluster_percent_heatmap
})

display(
    cluster_summary_heatmap
)


# ============================================================
# 10. OPTIONAL:
# REMOVE SINGLETONS FOR THE CROSS-SPECIES ANALYSIS
# ============================================================

cluster_taxonomy_heatmap_non_singleton = (
    cluster_taxonomy_heatmap[
        cluster_taxonomy_heatmap['n_peptides'] > 1
    ]
    .copy()
)

cluster_counts_non_singleton = (
    cluster_taxonomy_heatmap_non_singleton[
        'cluster_type'
    ]
    .value_counts()
    .reindex(
        [
            'Within species',
            'Cross-species\nwithin genus',
            'Cross-genus\nwithin family',
            'Cross-family'
        ]
    )
    .fillna(0)
)

cluster_percent_non_singleton = (
    100
    * cluster_counts_non_singleton
    / cluster_counts_non_singleton.sum()
)

cluster_summary_non_singleton = pd.DataFrame({
    'n_clusters': cluster_counts_non_singleton,
    'percent': cluster_percent_non_singleton
})

display(
    cluster_summary_non_singleton
)


# ============================================================
# 11. EXAMPLE CROSS-SPECIES CLUSTER
# ============================================================

example_clusters_heatmap = (
    cluster_taxonomy_heatmap[
        (cluster_taxonomy_heatmap['n_species'] > 1) &
        (cluster_taxonomy_heatmap['n_genera'] == 1)
    ]
    .sort_values(
        ['n_species', 'n_peptides'],
        ascending=False
    )
    .copy()
)

# Fallback: any cross-species cluster
if len(example_clusters_heatmap) == 0:

    example_clusters_heatmap = (
        cluster_taxonomy_heatmap[
            cluster_taxonomy_heatmap['n_species'] > 1
        ]
        .sort_values(
            ['n_species', 'n_peptides'],
            ascending=False
        )
        .copy()
    )


if len(example_clusters_heatmap) > 0:

    example_cluster_id_heatmap = (
        example_clusters_heatmap
        .iloc[0]['cluster_id']
    )

    example_cluster_heatmap = (
        hits_heatmap_z[
            hits_heatmap_z['cluster_id']
            == example_cluster_id_heatmap
        ]
        [
            [
                'peptide',
                'sequence',
                'species',
                'genus',
                'family',
                'protein_name'
            ]
        ]
        .drop_duplicates()
        .sort_values(
            ['species', 'peptide']
        )
        .reset_index(drop=True)
    )

    print(
        "Example cross-species cluster:",
        example_cluster_id_heatmap
    )

    display(
        example_cluster_heatmap
    )

else:

    print(
        "No cross-species clusters found "
        "among the heatmap species."
    )


# ============================================================
# 12. SAVE OBJECTS FOR DOWNSTREAM ANALYSIS
# ============================================================

hits_heatmap_z.to_csv(
    'hits_heatmap_species_with_sequence_clusters.csv',
    index=False
)

cluster_taxonomy_heatmap.to_csv(
    'cluster_taxonomy_heatmap_species.csv',
    index=False
)
```

```python
# ============================================================
# PLOTTING FUNCTION:
# SEQUENCE-LINKED PEPTIDE CLUSTER COMPOSITION
#
# Uses:
#   cluster_taxonomy_heatmap
#
# Shows:
#   number and percentage of non-singleton clusters spanning
#   increasing levels of viral taxonomy.
# ============================================================

def plot_peptide_cluster_composition(
    cluster_taxonomy_heatmap,
    figsize=(10, 6),
    exclude_singletons=True,
    title='Sequence relationships among high-specificity viral peptides'
):

    df = cluster_taxonomy_heatmap.copy()

    # --------------------------------------------------------
    # Classify clusters
    # --------------------------------------------------------

    def classify_cluster(row):

        if row['n_peptides'] == 1:
            return 'Singleton'

        elif row['n_species'] == 1:
            return 'Within species'

        elif row['n_genera'] == 1:
            return 'Cross-species\nwithin genus'

        elif row['n_families'] == 1:
            return 'Cross-genus\nwithin family'

        else:
            return 'Cross-family'

    df['cluster_type'] = (
        df.apply(
            classify_cluster,
            axis=1
        )
    )

    # --------------------------------------------------------
    # Optionally exclude singleton clusters
    # --------------------------------------------------------

    if exclude_singletons:

        df = df[
            df['cluster_type'] != 'Singleton'
        ].copy()

        cluster_order = [
            'Within species',
            'Cross-species\nwithin genus',
            'Cross-genus\nwithin family',
            'Cross-family'
        ]

    else:

        cluster_order = [
            'Singleton',
            'Within species',
            'Cross-species\nwithin genus',
            'Cross-genus\nwithin family',
            'Cross-family'
        ]

    # --------------------------------------------------------
    # Count
    # --------------------------------------------------------

    cluster_counts = (
        df['cluster_type']
        .value_counts()
        .reindex(
            cluster_order
        )
        .fillna(0)
    )

    # --------------------------------------------------------
    # Percentage
    # --------------------------------------------------------

    total = cluster_counts.sum()

    if total == 0:
        raise ValueError(
            "No peptide clusters available for plotting."
        )

    cluster_percent = (
        100
        * cluster_counts
        / total
    )

    # --------------------------------------------------------
    # Plot
    # --------------------------------------------------------

    fig, ax = plt.subplots(
        figsize=figsize
    )

    x = np.arange(
        len(cluster_counts)
    )

    bars = ax.bar(
        x,
        cluster_counts.values,
        width=0.7
    )

    ax.set_xticks(x)

    ax.set_xticklabels(
        cluster_counts.index,
        fontsize=11
    )

    ax.set_ylabel(
        'Number of peptide clusters',
        fontsize=14
    )

    ax.set_xlabel('')

    ax.set_title(
        title,
        fontsize=16,
        pad=15
    )

    # --------------------------------------------------------
    # Labels above bars
    # --------------------------------------------------------

    for i, bar in enumerate(bars):

        height = bar.get_height()

        if height > 0:

            ax.text(
                bar.get_x()
                + bar.get_width() / 2,
                height,
                (
                    f'{int(cluster_counts.iloc[i])}\n'
                    f'({cluster_percent.iloc[i]:.1f}%)'
                ),
                ha='center',
                va='bottom',
                fontsize=11
            )

    # --------------------------------------------------------
    # Clean up
    # --------------------------------------------------------

    sns.despine(
        ax=ax
    )

    ax.tick_params(
        axis='y',
        labelsize=11
    )

    plt.tight_layout()

    return (
        fig,
        ax,
        pd.DataFrame({
            'n_clusters': cluster_counts,
            'percent': cluster_percent
        })
    )
```

```python
# ============================================================
# PANEL A: SPECIES-LEVEL SEQUENCE-LINKAGE NETWORK
#
# Nodes:
#   viral species shown in the heatmap
#
# Node size:
#   number of SL individuals seroreactive to that species
#
# Edges:
#   number of sequence-linked peptide clusters containing
#   peptides from BOTH species
#
# Node color:
#   viral family
#
# Only cross-species sequence-linked clusters are used
# for edges.
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
from collections import defaultdict


# ============================================================
# 1. SPECIES TO INCLUDE
# ============================================================

network_species = set(
    species_order_reduced
)

print(
    f"Species in network: {len(network_species)}"
)


# ============================================================
# 2. RESTRICT PEPTIDES TO HEATMAP SPECIES
# ============================================================

network_hits = (
    hits_heatmap_z[
        hits_heatmap_z['species'].isin(
            network_species
        )
    ]
    .copy()
)


# ============================================================
# 3. SPECIES-LEVEL SEROPREVALENCE
# ============================================================

species_prevalence_network = (
    species_individual[
        (species_individual['cohort'] == 'SL') &
        (species_individual['species'].isin(network_species))
    ]
    .groupby('species')
    .agg(
        n_seroreactive=('seroreactive', 'sum'),
        fraction_seroreactive=('seroreactive', 'mean')
    )
    .reset_index()
)


# Add taxonomy
species_prevalence_network = (
    species_prevalence_network
    .merge(
        species_taxonomy,
        on='species',
        how='left'
    )
)


# ============================================================
# 4. IDENTIFY WHICH SPECIES OCCUR IN EACH CLUSTER
# ============================================================

cluster_species_network = (
    network_hits[
        [
            'cluster_id',
            'species'
        ]
    ]
    .drop_duplicates()
    .groupby('cluster_id')['species']
    .apply(list)
)


# ============================================================
# 5. BUILD SPECIES-SPECIES EDGE TABLE
#
# Every pair of species occurring in the same sequence-linked
# peptide cluster receives one edge contribution.
# ============================================================

species_edge_counts = defaultdict(int)

for cluster_id, species_list in (
    cluster_species_network.items()
):

    # Only clusters spanning >=2 species
    if len(species_list) < 2:
        continue

    # Unique species within cluster
    species_list = sorted(
        set(species_list)
    )

    for species_a, species_b in combinations(
        species_list,
        2
    ):

        species_edge_counts[
            (species_a, species_b)
        ] += 1


# Convert to dataframe
species_edges = pd.DataFrame(
    [
        {
            'species_a': species_a,
            'species_b': species_b,
            'n_shared_clusters': n_clusters
        }
        for (
            species_a,
            species_b
        ), n_clusters
        in species_edge_counts.items()
    ]
)


# ============================================================
# 6. BUILD GRAPH
# ============================================================

G_species = nx.Graph()


# Add nodes
for _, row in species_prevalence_network.iterrows():

    G_species.add_node(
        row['species'],
        family=row['family'],
        genus=row['genus'],
        n_seroreactive=row['n_seroreactive'],
        fraction_seroreactive=row['fraction_seroreactive']
    )


# Add edges
for _, row in species_edges.iterrows():

    G_species.add_edge(
        row['species_a'],
        row['species_b'],
        weight=row['n_shared_clusters']
    )


# ============================================================
# 7. KEEP ONLY CONNECTED SPECIES
#
# This removes species with no sequence-linked relationship
# to another displayed species.
# ============================================================

connected_species = [
    node
    for node, degree
    in G_species.degree()
    if degree > 0
]

G_species_connected = (
    G_species.subgraph(
        connected_species
    )
    .copy()
)

print(
    f"Connected species: "
    f"{G_species_connected.number_of_nodes()}"
)

print(
    f"Species-species edges: "
    f"{G_species_connected.number_of_edges()}"
)


# ============================================================
# 8. LAYOUT
# ============================================================

pos = nx.spring_layout(
    G_species_connected,
    weight='weight',
    seed=42,
    k=1.5
)


# ============================================================
# 9. NODE SIZE
# ============================================================

node_sizes = [
    300
    + 2500
    * G_species_connected.nodes[node][
        'fraction_seroreactive'
    ]
    for node in G_species_connected.nodes()
]


# ============================================================
# 10. EDGE WIDTH
# ============================================================

edge_widths = [
    1
    + 2
    * G_species_connected[u][v]['weight']
    for u, v in G_species_connected.edges()
]


# ============================================================
# 11. FAMILY COLORS
# ============================================================

families = sorted(
    set(
        nx.get_node_attributes(
            G_species_connected,
            'family'
        ).values()
    )
)

family_to_color = {
    family: plt.cm.tab20(i % 20)
    for i, family in enumerate(families)
}


node_colors = [
    family_to_color[
        G_species_connected.nodes[node]['family']
    ]
    for node in G_species_connected.nodes()
]


# ============================================================
# 12. PLOT
# ============================================================

fig, ax = plt.subplots(
    figsize=(14, 11)
)

nx.draw_networkx_edges(
    G_species_connected,
    pos,
    ax=ax,
    width=edge_widths,
    alpha=0.5
)

nx.draw_networkx_nodes(
    G_species_connected,
    pos,
    ax=ax,
    node_size=node_sizes,
    node_color=node_colors,
    edgecolors='black',
    linewidths=1
)

# Labels
nx.draw_networkx_labels(
    G_species_connected,
    pos,
    ax=ax,
    font_size=9
)

ax.set_title(
    'Sequence-linked relationships among seroreactive viral species',
    fontsize=16,
    pad=15
)

ax.axis('off')

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# BUILD PEPTIDE-LEVEL GRAPH
# ============================================================

G_peptide = nx.Graph()


# Add peptide nodes
for _, row in example_peptides.iterrows():

    G_peptide.add_node(
        row['peptide'],
        species=row['species'],
        genus=row['genus'],
        family=row['family'],
        protein=row['protein_name'],
        sequence=row['sequence']
    )


# ------------------------------------------------------------
# Connect peptides sharing an exact 8-aa k-mer
# ------------------------------------------------------------

K = 8


def get_kmers(
    sequence,
    k=K
):

    sequence = str(
        sequence
    ).upper()

    return {
        sequence[i:i+k]
        for i in range(
            len(sequence) - k + 1
        )
    }


kmer_to_peptides_example = defaultdict(list)

for _, row in example_peptides.iterrows():

    for kmer in get_kmers(
        row['sequence']
    ):

        kmer_to_peptides_example[kmer].append(
            row['peptide']
        )


# Add edges
for kmer, peptides in (
    kmer_to_peptides_example.items()
):

    if len(peptides) > 1:

        for peptide_a, peptide_b in combinations(
            peptides,
            2
        ):

            G_peptide.add_edge(
                peptide_a,
                peptide_b,
                kmer=kmer
            )


print(
    "Peptides:",
    G_peptide.number_of_nodes()
)

print(
    "Edges:",
    G_peptide.number_of_edges()
)
```

```python
import numpy as np

hbdb_stats = (
    master[master['category'] == 'HBDB']
    .groupby(['level', 'value'], observed=True)['sum_rpk']
    .agg(hbdb_mean = 'mean', hbdb_std='std', hbdb_max='max')
    .reset_index()
)

hbdb_stats['threshold_rpk_1sd'] = hbdb_stats['hbdb_mean'] + hbdb_stats['hbdb_std']
hbdb_stats['threshold_rpk_max'] = hbdb_stats['hbdb_max']

master = master.merge(
    hbdb_stats[['level', 'value', 'threshold_rpk_1sd', 'threshold_rpk_max']], 
    on=['level', 'value'], how='left'
)

#classify with an explicit "undetermined" bucket for missing reference data

master['seropositive_rpk_1sd'] = np.select(
    [master['threshold_rpk_1sd'].isna(), master['sum_rpk'] > master['threshold_rpk_1sd']], 
    ['Undetermined', 'Positive'], 
    default='Negative'
)

master['seropositive_rpk_max'] = np.select(
    [master['threshold_rpk_max'].isna(), master['sum_rpk'] > master['threshold_rpk_max']], 
    ['Undetermined', 'Positive'], 
    default='Negative'
)
```

```python
# pip install biopython  (if not already installed)
from Bio.Align import PairwiseAligner, substitution_matrices
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

TARGET_SPECIES = 'Reptarenavirus sp.'   # the "suspicious" non-local species
COMPARISON_RANK = 'family'                    # 'family' (broader) or 'genus' (narrower, faster)

aligner = PairwiseAligner()
aligner.substitution_matrix = substitution_matrices.load("BLOSUM62")
aligner.mode = 'global'
aligner.open_gap_score = -10
aligner.extend_gap_score = -0.5

def percent_identity(seq1, seq2):
    alignment = aligner.align(seq1, seq2)[0]
    s1, s2 = str(alignment[0]), str(alignment[1])
    matches = sum(a == b and a != '-' for a, b in zip(s1, s2))
    return 100 * matches / len(s1)

# --- peptide sets ---
target_family = long.loc[long['species'] == TARGET_SPECIES, COMPARISON_RANK].iloc[0]

target_peptides = (
    long[(long['species'] == TARGET_SPECIES)]
    [['peptide', 'sequence']]
    .drop_duplicates('peptide')
)

comparison_peptides = (
    long[(long[COMPARISON_RANK] == target_family) & (long['species'] != TARGET_SPECIES)]
    [['peptide', 'sequence', 'species']]
    .drop_duplicates('peptide')
)

print(f"{len(target_peptides)} target peptides x {len(comparison_peptides)} comparison peptides "
      f"= {len(target_peptides) * len(comparison_peptides)} alignments")
records = []
for _, t in target_peptides.iterrows():
    for _, c in comparison_peptides.iterrows():
        records.append({
            'target_peptide': t['peptide'],
            'comparison_peptide': c['peptide'],
            'comparison_species': c['species'],
            'pct_identity': percent_identity(t['sequence'], c['sequence']),
        })

identity_long = pd.DataFrame(records)

species_summary = (
    identity_long.groupby(['target_peptide', 'comparison_species'])['pct_identity']
    .max()
    .reset_index()
    .pivot(index='target_peptide', columns='comparison_species', values='pct_identity')
)
```

### making a peptide-level summary lookup table separate from master, which is taxon - level 
includes a fast sequence lookup and a plotting helper, all built off long since it already has everything we need (peptide, taxonomy, category, rpk, pass_z_3)

```python
def peptide_lookup_table(taxon_value, level='species', cohort=None, min_prevalence=0.00):
    subset = long[long[level] == taxon_value]
    if cohort is not None: 
        subset = subset[subset['cohort'] == cohort]
    

    stats = (
        subset.groupby('peptide', observed=True)
        .agg(
            n_individuals=('individual', 'nunique'), 
            n_hits=('pass_z_3', 'sum'), 
            mean_rpk=('rpk', 'mean'), 
            median_rpk=('rpk', 'median'), 
            sum_rpk=('rpk', 'sum'), 
            max_rpk=('rpk', 'max')
        )
        .reset_index()
    )
    stats['prevalence'] = stats['n_hits'] / stats['n_individuals']

    # static per-peptide metadata (one row per peptide, safe to merge back in)
    meta = subset.drop_duplicates('peptide')[
        ['peptide', 'sequence', 'protein_name', 'species', 'genus', 'family']
    ]
    stats = stats.merge(meta, on='peptide', how='left')

    return stats[stats['prevalence'] >= min_prevalence]
```

### helper code to identify top and bottom responders for peptides

```python
def top_bottom_individuals_per_peptide(peptide_ids, long_df=long, n=40):
    rows = []
    for peptide in peptide_ids:
        subset = long_df[long_df['peptide'] == peptide][
            ['peptide', 'individual', 'category', 'rpk', 'pass_z_3']
        ].sort_values('rpk', ascending=False)

        top = subset.head(n).copy()
        top['rank_group'] = 'top'

        bottom = subset.tail(n).sort_values('rpk', ascending=True).copy()
        bottom['rank_group'] = 'bottom'

        rows.append(pd.concat([top, bottom]))

    return pd.concat(rows, ignore_index=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np


def species_boxplot(data,  metadata, species:str, agg = "mean", dtype = "z_score"):
    """
    Parameters:
    - data: Pandas DataFrame containing the z score data 
    - x_var: column name 

    """
    species_peps = metadata[metadata['species'] == species]
    df_subset = data.loc[species_peps.index]
    
    #define group Sierra Leone (SL) vs HBDB
    sl_cols = [c for c in df_subset.columns if c.startswith(("C", "S"))]
    hbdb_cols = [c for c in df_subset.columns if c.startswith("HBDB")]

    #with the agg function, each individual (column) gets one value. returns a Series, so a single vector

    if agg == 'mean':
        sl_summary = df_subset[sl_cols].mean(axis=0, skipna=True)
        hbdb_summary=df_subset[hbdb_cols].mean(axis=0, skipna=True)

    elif agg == 'median':
        sl_summary = df_subset[sl_cols].median(axis=0, skipna=True)
        hbdb_summary=df_subset[hbdb_cols].median(axis=0, skipna=True)

    elif agg == 'max':
        sl_summary = df_subset[sl_cols].max(axis=0, skipna=True)
        hbdb_summary=df_subset[hbdb_cols].max(axis=0, skipna=True)

    elif agg == 'sum':
        sl_summary = df_subset[sl_cols].sum(axis=0, skipna=True)
        hbdb_summary=df_subset[hbdb_cols].sum(axis=0, skipna=True)

    else:
        raise ValueError("agg must be one of: 'mean', 'median' 'max', 'sum'")
 
    plot_df = pd.DataFrame({
        "value": pd.concat([sl_summary, hbdb_summary], axis=0).values,
        "group": ["SL"] * len(sl_summary.index)+ ["HBDB"] * len(hbdb_summary.index),
        "individual": list(sl_summary.index) +list(hbdb_summary.index)
    })

    
    print(df_subset.columns.tolist())
    
    sns.boxplot(data = plot_df, x = "group", y = "value")
    sns.stripplot(data=plot_df, x="group", y="value", color="black", alpha=0.6, jitter=True)
    plt.title(f"{species} peptides \n {agg} enrichment per individual", fontsize=13)
    plt.ylabel(f'{agg} {dtype} score')

    plt.savefig(f'./.png/{species}_max_{dtype}_enrichment_sl_hbdb_boxplot.png', dpi=300, bbox_inches='tight')
    plt.show()


```

# Sensitivity analysis for each peptide -> choosing the right threshold

```python
import numpy as np
import pandas as pd

#parameters to tune
thresholds = np.arange(3.0, 50.0, 1.0)   # candidate z-score thresholds for each peptide
max_healthy_positive_rate = 0.01        # 0.00 = no healthy positives allowed; try 0.01 if too strict -> 1% is allowed

# build peptide to species map
species_map = (
    peps_metadata[['peptide', 'species']]
    .drop_duplicates()
    .set_index('peptide')['species']
)

# Keep only peptides present in all three objects
common_peps = df_SL.index.intersection(df_hbdb.index).intersection(species_map.index)

df_SL2 = df_SL.loc[common_peps].copy()
df_hbdb2 = df_hbdb.loc[common_peps].copy()
species_map2 = species_map.loc[common_peps]

#choose a right threshold for each peptide using HC 
peptide_thresholds = {}
peptide_summary = []

for pep in common_peps:
    diseased_row = df_SL2.loc[pep]
    healthy_row = df_hbdb2.loc[pep]

    chosen_t = None
    chosen_hpr = None
    chosen_dpr = None

    # choose the lowest threshold that keeps healthy positive rate <= target
    for t in thresholds:
        healthy_pos_rate = (healthy_row > t).mean()
        diseased_pos_rate = (diseased_row > t).mean()

        if healthy_pos_rate <= max_healthy_positive_rate:
            chosen_t = t
            chosen_hpr = healthy_pos_rate
            chosen_dpr = diseased_pos_rate
            break

    # if nothing meets the healthy constraint, fall back to the highest threshold
    if chosen_t is None:
        chosen_t = thresholds[-1]
        chosen_hpr = (healthy_row > chosen_t).mean()
        chosen_dpr = (diseased_row > chosen_t).mean()

    peptide_thresholds[pep] = chosen_t
    peptide_summary.append({
        'peptide': pep,
        'species': species_map2.loc[pep],
        'threshold': chosen_t,
        'healthy_positive_rate': chosen_hpr,
        'exposed_positive_rate': chosen_dpr,
        'delta': chosen_dpr - chosen_hpr
    })

peptide_thresholds_df = pd.DataFrame(peptide_summary)

#convert each peptide into positive/negative calls using its own threshold
thr_vec = pd.Series(peptide_thresholds).reindex(df_SL2.index).to_numpy()[:, None]
sl_pos = (df_SL2.to_numpy() > thr_vec)      # shape: peptides x diseased samples
hb_pos = (df_hbdb2.to_numpy() > thr_vec)     # shape: peptides x healthy samples


# build sample-level species reactivity table 

species_sample_rows = []

species_to_peptides = species_map2.groupby(species_map2).groups

diseased_samples = df_SL2.columns
healthy_samples = df_hbdb2.columns

for species, peps in species_to_peptides.items():
    peps = list(peps)
    idx = df_SL2.index.get_indexer(peps)
    idx = idx[idx >= 0]

    if len(idx) == 0:
        continue

    n_peptides = len(idx)

    # counts per sample
    sl_counts = sl_pos[idx, :].sum(axis=0)
    hb_counts = hb_pos[idx, :].sum(axis=0)

    # diseased rows
    for sample, n_reactive in zip(diseased_samples, sl_counts):
        species_sample_rows.append({
            'sample': sample,
            'group': 'diseased',
            'species': species,
            'n_peptides': n_peptides,
            'n_reactive_peptides': int(n_reactive),
            'fraction_reactive_peptides': n_reactive / n_peptides
        })

    # healthy rows
    for sample, n_reactive in zip(healthy_samples, hb_counts):
        species_sample_rows.append({
            'sample': sample,
            'group': 'healthy',
            'species': species,
            'n_peptides': n_peptides,
            'n_reactive_peptides': int(n_reactive),
            'fraction_reactive_peptides': n_reactive / n_peptides
        })

species_sample_df = pd.DataFrame(species_sample_rows)

# Summarize species-level enrichment
species_summary = (
    species_sample_df
    .groupby('species', as_index=False)
    .agg(
        n_peptides=('n_peptides', 'first'),
        mean_reactive_diseased=('n_reactive_peptides', lambda x: x[species_sample_df.loc[x.index, 'group'].eq('diseased')].mean()),
        mean_reactive_healthy=('n_reactive_peptides', lambda x: x[species_sample_df.loc[x.index, 'group'].eq('healthy')].mean()),
        median_reactive_diseased=('n_reactive_peptides', lambda x: x[species_sample_df.loc[x.index, 'group'].eq('diseased')].median()),
        median_reactive_healthy=('n_reactive_peptides', lambda x: x[species_sample_df.loc[x.index, 'group'].eq('healthy')].median()),
        mean_fraction_diseased=('fraction_reactive_peptides', lambda x: x[species_sample_df.loc[x.index, 'group'].eq('diseased')].mean()),
        mean_fraction_healthy=('fraction_reactive_peptides', lambda x: x[species_sample_df.loc[x.index, 'group'].eq('healthy')].mean()),
    )
)

species_summary['diff_mean_fraction'] = (
    species_summary['mean_fraction_diseased'] - species_summary['mean_fraction_healthy']
)

species_summary['fold_enrichment'] = (
    species_summary['mean_fraction_diseased'] / (species_summary['mean_fraction_healthy'] + 1e-9)
)

species_summary = species_summary.sort_values(
    ['diff_mean_fraction', 'mean_fraction_diseased'],
    ascending=[False, False]
)

display(species_summary.head(50))
```

```python
import numpy as np 
import pandas as pd

#build peptide to species mapping
species_map = (
    peps_metadata[['peptide', 'species']]
    .drop_duplicates()
    .set_index('peptide')['species']
)

#make sure peptide names match the z-score matrix index
common_peps = df_SL.index.intersection(species_map.index)
df_SL2 = df_SL.loc[common_peps].copy()
df_hbdb2 = df_hbdb.loc[df_hbdb.index.intersection(species_map.index)].copy()
species_map2 = species_map.loc[common_peps]

#Threshold grid to test 
thresholds = np.arange(3.0, 50.0, 1.0) #start, stop, step

results = []

#loop over species
for species, peptides in species_map2.groupby(species_map2):
    peptides = peptides.index.tolist()

    sl_sub = df_SL2.loc[peptides]
    hb_sub = df_hbdb2.loc[peptides]

    for t in thresholds:
        #sample-level seropositivity: 
        #a sample is seropositive if ANY peptide in that species has z > threshold
        diseased_pos = (sl_sub > t).any(axis=0)
        healthy_pos = (hb_sub > t).any(axis=0)

        sensitivity = diseased_pos.mean() #TP / all diseased
        specificity = 1 - healthy_pos.mean()
        youden_j = sensitivity + specificity - 1
        balanced_accuracy = (sensitivity + specificity) / 2 

        results.append({
            'species': species,
            'threshold': t, 
            'n_peptides': len(peptides), 
            'sensitivity': sensitivity, 
            'specificity': specificity, 
            'youden_j': youden_j, 
            'balanced_accuracy': balanced_accuracy, 
            'SL_positive_rate':diseased_pos.mean(), 
            'healthy_positive_rate': healthy_pos.mean()
        })

results_df = pd.DataFrame(results)
```

# LASV survivor durability — Figure 3 Panels B–D

Focused analysis of confirmed Lassa fever survivors with a known date of discharge (DoDis).

**Panel B:** NP ELISA, GP ELISA, and neutralization versus years since discharge.  
**Panel C:** NP and GPC high-specificity PhIP-Seq peptide breadth versus years since discharge.  
**Panel D:** adjusted time-since-discharge effect estimates with 95% confidence intervals.

All adjusted models include **age and sex** as covariates. The first/baseline sample per survivor is used so each participant contributes once.


```python
from pathlib import Path
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf
import statsmodels.api as sm
from scipy.stats import spearmanr
from statsmodels.stats.multitest import multipletests

warnings.filterwarnings('ignore', category=FutureWarning)

# ---------- paths ----------
META_FILE = Path('./metadata_files/master_sample_metadata_111226.csv')
LONG_FILE = Path('./output_files/long_spec_0_98_111226.csv')
OUTDIR = Path('./output_figures/Fig3_DoDis')
OUTDIR.mkdir(parents=True, exist_ok=True)

# ---------- publication style ----------
SURVIVOR_COLOR = '#D55E00'  # Okabe–Ito vermillion
plt.rcParams.update({
    'font.size': 9,
    'axes.labelsize': 10,
    'axes.titlesize': 10,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
    'legend.fontsize': 9,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'svg.fonttype': 'none',
    'pdf.fonttype': 42,
})

```

## 1. Build the survivor DoDis analysis cohort


```python
sample_meta = pd.read_csv(META_FILE)
df = pd.read_csv(LONG_FILE)

sample_meta['collection_date'] = pd.to_datetime(sample_meta['collection_date'], errors='coerce')
sample_meta['discharge_date'] = pd.to_datetime(sample_meta['discharge_date'], errors='coerce')

# Recalculate DoDis from dates as a QC safeguard.
sample_meta['days_since_discharge_calc'] = (
    sample_meta['collection_date'] - sample_meta['discharge_date']
).dt.days
sample_meta['years_since_discharge_calc'] = sample_meta['days_since_discharge_calc'] / 365.25

# Confirmed survivors, known DoDis, first available sample per participant.
survivors = (
    sample_meta.loc[
        (sample_meta['Category'] == 'S') &
        sample_meta['discharge_date'].notna() &
        sample_meta['collection_date'].notna()
    ]
    .sort_values(['participant_id', 'collection_date'])
    .drop_duplicates('participant_id', keep='first')
    .copy()
)

# Use recalculated value so collection/discharge dates are the source of truth.
survivors['years_since_discharge'] = survivors['years_since_discharge_calc']

# Numeric outcomes/covariates.
for col in ['NP_Calc_Conc', 'GP_Calc_Conc', 'Neut_Percent', 'age_years']:
    survivors[col] = pd.to_numeric(survivors[col], errors='coerce')

# Standardize sex labels to a simple categorical variable; inspect before modeling.
survivors['sex_model'] = survivors['sex'].astype('string').str.strip()

print('Survivors with known DoDis:', len(survivors))
print('Years since discharge:', survivors['years_since_discharge'].describe()[['min','25%','50%','75%','max']])
print()
print('Sex:')
print(survivors['sex_model'].value_counts(dropna=False))
print()
print('Age available:', survivors['age_years'].notna().sum())
print('NP ELISA available:', survivors['NP_Calc_Conc'].notna().sum())
print('GP ELISA available:', survivors['GP_Calc_Conc'].notna().sum())
print('Neutralization available:', survivors['Neut_Percent'].notna().sum())

```

## 2. Calculate antigen-specific PhIP-Seq breadth


```python
baseline_ids = set(survivors['sample_id'])

lasv = df.loc[
    df['individual'].isin(baseline_ids) &
    df['species'].eq('Mammarenavirus lassaense')
].copy()

# Number of reactive high-specificity peptides per protein.
breadth = (
    lasv.groupby(['individual', 'protein_std'])['reactive']
        .sum()
        .unstack(fill_value=0)
        .reset_index()
        .rename(columns={
            'nucleoprotein': 'NP_breadth',
            'glycoprotein': 'GPC_breadth',
            'polymerase': 'L_breadth',
            'z protein': 'Z_breadth',
        })
)

analysis = survivors.merge(
    breadth,
    left_on='sample_id',
    right_on='individual',
    how='left',
    validate='one_to_one'
)

for col in ['NP_breadth', 'GPC_breadth', 'L_breadth', 'Z_breadth']:
    if col not in analysis.columns:
        analysis[col] = 0
    analysis[col] = analysis[col].fillna(0).astype(int)

print('Analysis N:', len(analysis))
display(analysis[['participant_id','years_since_discharge','NP_Calc_Conc','GP_Calc_Conc','Neut_Percent','NP_breadth','GPC_breadth']].head())

```

## 3. Fit adjusted models

The adjusted models are:

- `log1p(NP ELISA) ~ years since discharge + age + sex`
- `log1p(GP ELISA) ~ years since discharge + age + sex`
- `Neutralization (%) ~ years since discharge + age + sex`
- `NP breadth ~ years since discharge + age + sex` (negative binomial)
- `GPC breadth ~ years since discharge + age + sex` (negative binomial)

The forest plot reports the time effect over **10 years**, which is easier to interpret than a one-year coefficient.


```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf

from statsmodels.discrete.discrete_model import NegativeBinomial
from statsmodels.stats.multitest import multipletests
import patsy

# statsmodels/patsy does not handle pandas StringDtype reliably
analysis = analysis.copy()

analysis['sex_model'] = analysis['sex_model'].astype(object)

# make sure all modelled numeric variables really are numeric
numeric_cols = [
    'years_since_discharge',
    'age_years',
    'NP_Calc_Conc',
    'GP_Calc_Conc',
    'Neut_Percent',
    'NP_breadth',
    'GPC_breadth'
]

for col in numeric_cols:
    analysis[col] = pd.to_numeric(
        analysis[col],
        errors='coerce'
    )

    
def fit_adjusted_models(data):
    rows = []
    fitted = {}

    specs = [
        ('NP ELISA', 'NP_Calc_Conc', 'ols_log1p'),
        ('GP ELISA', 'GP_Calc_Conc', 'ols_log1p'),
        ('Neutralization', 'Neut_Percent', 'ols'),
        ('NP PhIP breadth', 'NP_breadth', 'neg_bin'),
        ('GPC PhIP breadth', 'GPC_breadth', 'neg_bin'),
    ]

    for label, outcome, model_type in specs:

        cols = [
            outcome,
            'years_since_discharge',
            'age_years',
            'sex_model'
        ]

        tmp = data[cols].dropna().copy()

        # -----------------------------------
        # ELISA: OLS on log1p concentration
        # -----------------------------------
        if model_type == 'ols_log1p':

            tmp['model_outcome'] = np.log1p(tmp[outcome])

            model = smf.ols(
                'model_outcome ~ years_since_discharge '
                '+ age_years + C(sex_model)',
                data=tmp
            ).fit()

        # -----------------------------------
        # Neutralization: ordinary OLS
        # -----------------------------------
        elif model_type == 'ols':

            model = smf.ols(
                f'{outcome} ~ years_since_discharge '
                '+ age_years + C(sex_model)',
                data=tmp
            ).fit()

        # -----------------------------------
        # PhIP breadth: Negative Binomial
        # with dispersion estimated from data
        # -----------------------------------
        elif model_type == 'neg_bin':

            formula = (
                f'{outcome} ~ years_since_discharge '
                '+ age_years + C(sex_model)'
            )

            y, X = patsy.dmatrices(
                formula,
                data=tmp,
                return_type='dataframe'
            )

            model = NegativeBinomial(
                y.iloc[:, 0],
                X
            ).fit(
                disp=False,
                maxiter=1000
            )

        # -----------------------------------
        # Extract time effect
        # -----------------------------------

        beta = model.params['years_since_discharge']

        ci = model.conf_int().loc['years_since_discharge']
        ci_low = ci.iloc[0]
        ci_high = ci.iloc[1]

        p = model.pvalues['years_since_discharge']

        # -----------------------------------
        # Convert to 10-year effect
        # -----------------------------------

        if model_type in ['ols_log1p', 'neg_bin']:

            effect10 = np.exp(beta * 10)
            low10 = np.exp(ci_low * 10)
            high10 = np.exp(ci_high * 10)

            percent_change10 = (
                effect10 - 1
            ) * 100

            if model_type == 'neg_bin':
                effect_type = 'rate ratio per 10 years'
            else:
                effect_type = (
                    'ratio of (ELISA concentration + 1) '
                    'per 10 years'
                )

        else:

            effect10 = beta * 10
            low10 = ci_low * 10
            high10 = ci_high * 10

            percent_change10 = np.nan

            effect_type = (
                'percentage-point change per 10 years'
            )

        rows.append({
            'outcome': label,
            'raw_outcome': outcome,
            'model': model_type,
            'n': len(tmp),

            'beta_per_year': beta,
            'ci_low_per_year': ci_low,
            'ci_high_per_year': ci_high,

            'p_time': p,

            'effect_10y': effect10,
            'effect_10y_low': low10,
            'effect_10y_high': high10,

            'percent_change_10y': percent_change10,

            'effect_type': effect_type
        })

        fitted[label] = model

    # -----------------------------------
    # Multiple-testing correction
    # -----------------------------------

    results = pd.DataFrame(rows)

    results['FDR_time'] = multipletests(
        results['p_time'],
        method='fdr_bh'
    )[1]

    return results, fitted


model_results, fitted_models = fit_adjusted_models(
    analysis
)

display(model_results)
```

## 4. Panel B — humoral waning

Raw observations are shown together with **predictions from the same age- and sex-adjusted model used for inference**. Prediction curves are standardized to the cohort median age and modal sex; shaded bands are 95% confidence intervals for the adjusted mean response.

Annotations report the adjusted effect of **10 additional years since discharge** and BH-FDR across the five primary durability outcomes.


```python
# ---------- adjusted prediction helpers ----------
PREDICTION_AGE = float(analysis['age_years'].median())
PREDICTION_SEX = analysis['sex_model'].mode(dropna=True).iloc[0]

def adjusted_annotation(label):
    r = model_results.set_index('outcome').loc[label]

    if r['model'] == 'ols':
        return (
            f"{r['effect_10y']:.1f} pp / decade\n"
            f"FDR = {r['FDR_time']:.3g}"
        )

    pct = (1 - r['effect_10y']) * 100
    return (
        f"{pct:.0f}% lower / decade\n"
        f"FDR = {r['FDR_time']:.3g}"
    )


def adjusted_prediction(label, data, n_grid=200):
    """
    Prediction from the exact fitted model, holding age at the cohort median
    and sex at the cohort modal category.

    Returns columns: years_since_discharge, mean, lower, upper.
    """
    model = fitted_models[label]

    x_min = float(data['years_since_discharge'].min())
    x_max = float(data['years_since_discharge'].max())
    grid = np.linspace(x_min, x_max, n_grid)

    new = pd.DataFrame({
        'years_since_discharge': grid,
        'age_years': PREDICTION_AGE,
        'sex_model': [PREDICTION_SEX] * n_grid,
    })

    model_type = model_results.set_index('outcome').loc[label, 'model']

    if model_type == 'ols_log1p':
        pred = model.get_prediction(new).summary_frame(alpha=0.05)

        # Model is on log1p(outcome); transform adjusted means and CI back
        # to the original concentration scale.
        mean = np.expm1(pred['mean'].to_numpy())
        lower = np.expm1(pred['mean_ci_lower'].to_numpy())
        upper = np.expm1(pred['mean_ci_upper'].to_numpy())

    elif model_type == 'ols':
        pred = model.get_prediction(new).summary_frame(alpha=0.05)
        mean = pred['mean'].to_numpy()
        lower = pred['mean_ci_lower'].to_numpy()
        upper = pred['mean_ci_upper'].to_numpy()

    elif model_type == 'neg_bin':
        # Recreate the exact design matrix used by the discrete NB model.
        # Explicit category levels keep Patsy columns aligned with the fit.
        sex_levels = sorted(data['sex_model'].dropna().astype(str).unique())
        new = new.copy()
        new['sex_model'] = pd.Categorical(
            new['sex_model'].astype(str),
            categories=sex_levels
        )

        formula_rhs = 'years_since_discharge + age_years + C(sex_model)'
        Xnew = patsy.dmatrix(
            formula_rhs,
            data=new,
            return_type='dataframe'
        )

        # Align columns to coefficient names, excluding the estimated
        # dispersion parameter alpha.
        coef_names = [c for c in model.params.index if c != 'alpha']
        Xnew = Xnew.reindex(columns=coef_names, fill_value=0.0)

        beta = model.params.loc[coef_names].to_numpy()
        cov = model.cov_params().loc[coef_names, coef_names].to_numpy()

        eta = Xnew.to_numpy() @ beta
        Xarr = Xnew.to_numpy()
        se_eta = np.sqrt(
            np.einsum('ij,jk,ik->i', Xarr, cov, Xarr)
        )

        mean = np.exp(eta)
        lower = np.exp(eta - 1.96 * se_eta)
        upper = np.exp(eta + 1.96 * se_eta)

    else:
        raise ValueError(f'Unknown model type: {model_type}')

    # Outcomes cannot be negative.
    if label != 'Neutralization':
        mean = np.clip(mean, 0, None)
        lower = np.clip(lower, 0, None)
        upper = np.clip(upper, 0, None)

    return pd.DataFrame({
        'years_since_discharge': grid,
        'mean': mean,
        'lower': lower,
        'upper': upper,
    })


def add_adjusted_curve(ax, label):
    pred = adjusted_prediction(label, analysis)

    ax.plot(
        pred['years_since_discharge'],
        pred['mean'],
        color=SURVIVOR_COLOR,
        lw=1.7,
        zorder=3
    )

    ax.fill_between(
        pred['years_since_discharge'],
        pred['lower'],
        pred['upper'],
        color=SURVIVOR_COLOR,
        alpha=0.13,
        linewidth=0,
        zorder=1
    )


panel_b_specs = [
    ('NP ELISA', 'NP_Calc_Conc', 'NP ELISA concentration'),
    ('GP ELISA', 'GP_Calc_Conc', 'GP ELISA concentration'),
    ('Neutralization', 'Neut_Percent', 'Neutralization (%)'),
]

fig, axes = plt.subplots(1, 3, figsize=(10.8, 3.2))

for ax, (label, col, ylabel) in zip(axes, panel_b_specs):
    tmp = analysis[['years_since_discharge', col]].dropna()

    ax.scatter(
        tmp['years_since_discharge'],
        tmp[col],
        s=20,
        alpha=0.62,
        color=SURVIVOR_COLOR,
        edgecolor='none',
        zorder=2
    )

    add_adjusted_curve(ax, label)

    ax.set_xlabel('Years since discharge')
    ax.set_ylabel(ylabel)
    ax.text(
        0.97, 0.96,
        adjusted_annotation(label),
        transform=ax.transAxes,
        ha='right',
        va='top',
        fontsize=8
    )

fig.suptitle(
    f'Adjusted curves standardized to age {PREDICTION_AGE:.0f} years and sex = {PREDICTION_SEX}',
    y=1.02,
    fontsize=8,
    color='0.35'
)
fig.tight_layout()
fig.savefig(OUTDIR / 'Fig3B_humoral_waning_adjusted.svg', bbox_inches='tight')
fig.savefig(OUTDIR / 'Fig3B_humoral_waning_adjusted.pdf', bbox_inches='tight')
plt.show()
```

## 5. Panel C — antigen-specific PhIP-Seq breadth waning

NP and GPC breadth are count outcomes and are modeled with **negative-binomial regression** because both are overdispersed. The fitted curves below are therefore negative-binomial model predictions—not linear regression lines—and remain non-negative by construction.


```python
panel_c_specs = [
    ('NP PhIP breadth', 'NP_breadth', 'NP high-specificity peptide breadth'),
    ('GPC PhIP breadth', 'GPC_breadth', 'GPC high-specificity peptide breadth'),
]

fig, axes = plt.subplots(1, 2, figsize=(7.4, 3.2))

for ax, (label, col, ylabel) in zip(axes, panel_c_specs):
    tmp = analysis[['years_since_discharge', col]].dropna()

    # Small deterministic vertical jitter only for display of integer counts.
    rng = np.random.default_rng(42)
    y_plot = tmp[col].astype(float) + rng.normal(0, 0.07, len(tmp))

    ax.scatter(
        tmp['years_since_discharge'],
        y_plot,
        s=20,
        alpha=0.62,
        color=SURVIVOR_COLOR,
        edgecolor='none',
        zorder=2
    )

    add_adjusted_curve(ax, label)

    ax.set_xlabel('Years since discharge')
    ax.set_ylabel(ylabel)
    ax.set_ylim(bottom=-0.5)
    ax.text(
        0.97, 0.96,
        adjusted_annotation(label),
        transform=ax.transAxes,
        ha='right',
        va='top',
        fontsize=8
    )

fig.suptitle(
    f'Adjusted curves standardized to age {PREDICTION_AGE:.0f} years and sex = {PREDICTION_SEX}',
    y=1.02,
    fontsize=8,
    color='0.35'
)
fig.tight_layout()
fig.savefig(OUTDIR / 'Fig3C_PhIP_breadth_waning_adjusted.svg', bbox_inches='tight')
fig.savefig(OUTDIR / 'Fig3C_PhIP_breadth_waning_adjusted.pdf', bbox_inches='tight')
plt.show()
```

## 6. Panel D — adjusted effect-size summary

Panel D summarizes the adjusted **10-year time effect** with 95% confidence intervals. Multiplicative outcomes are displayed as fold-change per decade (null = 1); neutralization is displayed separately as an absolute percentage-point change per decade (null = 0). This keeps incompatible effect scales visually distinct.


```python
ratio_labels = [
    'NP ELISA',
    'GP ELISA',
    'NP PhIP breadth',
    'GPC PhIP breadth'
]

ratio_df = (
    model_results
    .set_index('outcome')
    .loc[ratio_labels]
    .reset_index()
)

neut_df = model_results.loc[
    model_results['outcome'] == 'Neutralization'
].copy()

fig, axes = plt.subplots(
    1, 2,
    figsize=(9.0, 3.55),
    gridspec_kw={'width_ratios': [1.65, 1]}
)

# ---------- multiplicative outcomes ----------
ax = axes[0]
y = np.arange(len(ratio_df))[::-1]

ax.errorbar(
    ratio_df['effect_10y'],
    y,
    xerr=[
        ratio_df['effect_10y'] - ratio_df['effect_10y_low'],
        ratio_df['effect_10y_high'] - ratio_df['effect_10y']
    ],
    fmt='o',
    color=SURVIVOR_COLOR,
    ecolor='0.35',
    elinewidth=1.2,
    capsize=3,
    markersize=5,
    zorder=3
)

ax.axvline(1, color='0.65', ls='--', lw=1, zorder=0)
ax.set_yticks(y)
ax.set_yticklabels(ratio_df['outcome'])
ax.set_xlabel('Adjusted fold-change per decade')
ax.set_title('Binding and PhIP-Seq breadth', loc='left')

# Keep enough room for CI and annotation.
xmin = min(0.35, ratio_df['effect_10y_low'].min() - 0.03)
ax.set_xlim(xmin, 1.08)

for yi, (_, r) in zip(y, ratio_df.iterrows()):
    pct_lower = (1 - r['effect_10y']) * 100
    ax.text(
        1.065,
        yi,
        f"{pct_lower:.0f}% lower\nFDR {r['FDR_time']:.3f}",
        ha='right',
        va='center',
        fontsize=7.5
    )

# ---------- neutralization ----------
ax = axes[1]
r = neut_df.iloc[0]

ax.errorbar(
    r['effect_10y'],
    0,
    xerr=[
        [r['effect_10y'] - r['effect_10y_low']],
        [r['effect_10y_high'] - r['effect_10y']]
    ],
    fmt='o',
    color=SURVIVOR_COLOR,
    ecolor='0.35',
    elinewidth=1.2,
    capsize=3,
    markersize=5,
    zorder=3
)

ax.axvline(0, color='0.65', ls='--', lw=1, zorder=0)
ax.set_yticks([0])
ax.set_yticklabels(['Neutralization'])
ax.set_xlabel('Adjusted percentage-point change\nper decade')
ax.set_title('Functional activity', loc='left')

ax.text(
    0.97, 0.92,
    f"{r['effect_10y']:.1f} pp / decade\n"
    f"FDR {r['FDR_time']:.3f}",
    transform=ax.transAxes,
    ha='right',
    va='top',
    fontsize=7.5
)

fig.tight_layout(w_pad=2.5)
fig.savefig(OUTDIR / 'Fig3D_adjusted_effect_forest_improved.svg', bbox_inches='tight')
fig.savefig(OUTDIR / 'Fig3D_adjusted_effect_forest_improved.pdf', bbox_inches='tight')
plt.show()
```

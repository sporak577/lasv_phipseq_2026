```python
import pandas as pd

df= pd.read_csv('./output_files/long_spec_0_98_111226.csv')

top_peptides = pd.read_csv('./output_files/pep_hits_wo_taxa_to_remove_less_stringent_spec_0_98_111226.csv')

peps_metadata = pd.read_csv("./metadata_files/lassa_library_sequences_species_standardized_070726_ICTV_corrected.csv")
peps_metadata = peps_metadata.set_index('seq_id')
```

```python
#from long to wide
plot_df = df.pivot(index='peptide', columns='individual', values='z_score')

#subset peps metadata to pep hits
pep_hits_meta = peps_metadata.loc[plot_df.index]
```

### Looking at the overall peptide reactivity landscape 

```python
n_c = plot_df.filter(regex='C-').shape[1]
n_cs = plot_df.shape[1]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

#sort by genus
pep_hits_meta = pep_hits_meta.sort_values(by=['family', 'species'])

# sort families by size (largest first), then by species within each family
pep_hits_meta['family_n'] = pep_hits_meta.groupby('family')['family'].transform('size')
pep_hits_meta = pep_hits_meta.sort_values(
    by=['family_n', 'family', 'species'],
    ascending=[False, True, True],
).drop(columns='family_n')

plot_df = plot_df.reindex(pep_hits_meta.index)

genera = pep_hits_meta['family'].unique()   # now ordered largest -> smallest

plot_df = plot_df.reindex(pep_hits_meta.index)

#one color per genus
genera = pep_hits_meta['family'].unique() #order follows the sort

palette = sns.color_palette('tab20', n_colors=len(genera))
genus_lut = dict(zip(genera, palette)) #zip pairs up two lists item by item, dict turns them into key -> value

#.map(genus_lut) swaps each genus name in the column for its color in the dictionary. 
# result is Series of color with one entry per peptide, keeps same index as pep_hits_meta
#clustermap uses that index to match each color to its row in plot_df
#.Genus changes the Series' name from the column. purely cosmetic, as clustermap will write this name next to color strip
row_colors = pep_hits_meta['family'].map(genus_lut).rename('Family')

g = sns.clustermap(
    plot_df, 
    row_cluster=False, 
    col_cluster=False, 
    row_colors=row_colors, 
    colors_ratio=0.01,
    cmap='Blues', 
    yticklabels=False, 
    rasterized=True,
    figsize=(15, 10),
    vmin=3, vmax=10
)

# group dividers
for x in (n_c, n_cs):
    g.ax_heatmap.axvline(x, color="black", lw=0.6)

handles = [mpatches.Patch(color=genus_lut[gn], label=gn) for gn in genera]
g.ax_heatmap.legend(handles=handles, title='Family', loc='lower right', frameon=False, bbox_to_anchor=(-0.05, 0),
            fontsize=15, title_fontsize=16)

g.ax_cbar.tick_params(labelsize=15)

# group dividers (samples)
for x in (n_c, n_cs):
    g.ax_heatmap.axvline(x, color="black", lw=0.6)

# family dividers (peptides)
counts = pep_hits_meta['family'].value_counts().reindex(genera)
for y in np.cumsum(counts.values)[:-1]:
    g.ax_heatmap.axhline(y, color="black", lw=0.6, snap=True, antialiased=False)
    g.ax_row_colors.axhline(y, color="black", lw=0.6)   # also across the color strip


#plt.savefig('./svg_figures/peptide_hits_birdseye_view_by_family.svg', format='svg',bbox_inches='tight')
plt.show()
```

```python
# flag Lassa peptides
is_lassa = pep_hits_meta['species'].eq('Mammarenavirus lassaense')
print(is_lassa.sum(), 'Lassa peptides')   # check it matched

# two color strips: family + Lassa highlight
row_colors = pd.DataFrame({
    'Family': pep_hits_meta['family'].map(genus_lut),
    'Lassa':  is_lassa.map({True: 'black', False: 'white'}),
})

g = sns.clustermap(
    plot_df,
    row_cluster=False,
    col_cluster=False,
    rasterized=True,
    row_colors=row_colors,
    colors_ratio=0.01,
    cmap='Blues',
    yticklabels=False,
    figsize=(15, 10),
    vmin=3, vmax=10,
)

# group dividers (samples)
for x in (n_c, n_cs):
    g.ax_heatmap.axvline(x, color="black", lw=0.6)

# family dividers (peptides)
counts = pep_hits_meta['family'].value_counts().reindex(genera)
for y in np.cumsum(counts.values)[:-1]:
    g.ax_heatmap.axhline(y, color="black", lw=0.6, snap=True, antialiased=False)
    g.ax_row_colors.axhline(y, color="black", lw=0.6)

# outline the Lassa block on the heatmap
rows = np.flatnonzero(is_lassa.values)
start, end = rows.min(), rows.max() + 1
g.ax_heatmap.add_patch(mpatches.Rectangle(
    (0, start), plot_df.shape[1], end - start,
    fill=False, edgecolor='black', lw=2, clip_on=False,
))

# legend
handles = [mpatches.Patch(color=genus_lut[gn], label=gn) for gn in genera]
handles.append(mpatches.Patch(color='red', label='M. lassaense'))
g.ax_heatmap.legend(handles=handles, title='Family', loc='lower right', frameon=False,
                    bbox_to_anchor=(-0.05, 0), fontsize=15, title_fontsize=16)

g.ax_cbar.tick_params(labelsize=15)

#plt.savefig('./svg_figures/peptide_hits_birdseye_view_by_famil_lassa_peps.svg', bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from matplotlib.ticker import MultipleLocator, PercentFormatter

# SETTINGS

plt.rcParams.update({
    'font.size': 9.5, 'axes.labelsize': 10, 'axes.titlesize': 11,
    'xtick.labelsize': 9.5, 'ytick.labelsize': 9.5, 'legend.fontsize': 9,
    'axes.spines.top': False, 'axes.spines.right': False,
    'svg.fonttype': 'none', 'pdf.fonttype': 42,
})


BROAD_THRESHOLD = 0.20
N_COHORT = 397


TOP_VIRAL_FAMILIES = ['Phenuiviridae', 'Flaviviridae', 'Arenaviridae',
       'Hantaviridae']
# STYLE

# Other hemorrhagic-fever-associated families:
# matches the pale-pink language of the library schematic
VHF_COLOR = "#D98C8C"

# Subtle 20% reference
THRESHOLD_COLOR = "#A6A6A6"

# Main text / axes
TEXT_COLOR = "#111111"

# ---- add to SETTINGS / STYLE ----
SPECIES_COL = "species"   # <- change if your column is named differently

TOP_SPECIES = {
    # species: (short label, color, species-level seroprevalence)
    "Bandavirus dabieense":          ("SFTSV", "#C23B6E", 0.788413),  # raspberry
    "Bandavirus guertuense":         ("GTV",   "#E89A3C", 0.707809),  # orange
    "Orthoflavivirus denguei":       ("DENV",  "#2E3050", 0.637280),  # navy
    "Mammarenavirus lassaense":      ("LASV",  "#7F83C9", 0.536524),  # periwinkle
    "Orthohantavirus sinnombreense": ("SNV",   "#D98C8C", 0.526448),  # pale rose (your VHF pink)
}

BACKGROUND_COLOR = "#D4D4D4"   # lighter than the phage grey so colored dots pop

BACKGROUND_COLOR = "#C8C8C8"
SHOW_SPECIES_SEROPREV = True   # open diamond = % reactive to ≥1 peptide of the species

MARKER_DX = 0.00   # horizontal shift, in axis units (0.01 = 1 percentage point)
MARKER_DY = -0.25   # vertical shift; the y axis is inverted, so + moves it down

# 1. PREPARE DATA

# 1. PREPARE DATA  (add SPECIES_COL to the dropna; rest unchanged)

plot_df = pd.concat(
    [
        df.assign(family=family)
        for family, df in family_hits_sub_dfs.items()
    ],
    ignore_index=True
)

plot_df = plot_df.dropna(
    subset=["peptide", "family", "SL_prevalence", SPECIES_COL]
).copy()


plot_df["SL_prevalence"] = pd.to_numeric(
    plot_df["SL_prevalence"],
    errors="coerce"
)

plot_df = plot_df.dropna(
    subset=[
        "peptide",
        "family",
        "SL_prevalence"
    ]
).copy()


# 2. KEEP ONLY VHF-ASSOCIATED FAMILIES

vhf_df = plot_df[
    plot_df["family"].isin(TOP_VIRAL_FAMILIES)
].copy()


# 3. FAMILY SUMMARY

family_summary = (
    vhf_df
    .groupby("family")
    .agg(
        n_peptides=("peptide", "nunique"),
        max_sensitivity=("SL_prevalence", "max")
    )
)


# 4. ORDER FAMILIES
#
# Families are ordered by their most broadly recognized
# peptide.

family_order = (
    family_summary
    .sort_values(
        "max_sensitivity",
        ascending=False
    )
    .index
    .tolist()
)

print(
    family_summary.loc[family_order]
)


# 5. FIGURE


fig, ax = plt.subplots(
    figsize=(4.2, 3.2)
)

fig.patch.set_facecolor("white")
ax.set_facecolor("white")

rng = np.random.default_rng(42)


# 6. PLOT PEPTIDES


# 6. PLOT PEPTIDES (replaces the old loop)

for ypos, family in enumerate(family_order):

    sub = vhf_df[vhf_df["family"].eq(family)].copy()
    sub["y"] = ypos + rng.uniform(-0.14, 0.14, size=len(sub))

    is_top = sub[SPECIES_COL].isin(TOP_SPECIES)

    # Background: all other peptides in this family
    ax.scatter(
        sub.loc[~is_top, "SL_prevalence"], sub.loc[~is_top, "y"],
        s=12, color=BACKGROUND_COLOR, alpha=0.6,
        edgecolor="none", rasterized=True, zorder=2
    )

    # Foreground: highlighted species
    top_here = [s for s in TOP_SPECIES if s in set(sub[SPECIES_COL])]

    for k, species in enumerate(top_here):
        label, color, sp_prev = TOP_SPECIES[species]
        s_df = sub[sub[SPECIES_COL].eq(species)]

        ax.scatter(
            s_df["SL_prevalence"], s_df["y"],
            s=16, color=color, alpha=0.9,
            edgecolor="none",
            rasterized=True, zorder=3
        )
# 7. 20% COHORT-REACTIVITY REFERENCE

ax.axvline(
    BROAD_THRESHOLD,
    color=THRESHOLD_COLOR,
    linestyle=(0, (3, 3)),
    linewidth=0.7,
    zorder=1
)


# 8. FAMILY LABELS
ax.set_yticklabels(family_order, color=TEXT_COLOR)

ax.set_yticks(
    range(len(family_order))
)

ax.invert_yaxis()

ax.tick_params(
    axis="y",
    length=0,
    pad=10,
    colors=TEXT_COLOR
)

# 9. X AXIS

# Round maximum upward to nearest 10%

# 9. X AXIS: include the species markers in the range
x_candidates = [vhf_df["SL_prevalence"].max()]
if SHOW_SPECIES_SEROPREV:
    x_candidates.append(max(v[2] for v in TOP_SPECIES.values()))


xmax = np.ceil(max(x_candidates) * 10) / 10

ax.set_xlim(
    0,
    xmax
)

ax.xaxis.set_major_locator(
    MultipleLocator(0.1)
)

ax.xaxis.set_major_formatter(
    PercentFormatter(
        xmax=1,
        decimals=0
    )
)

ax.set_xlabel(
    "Individuals reactive to peptide",
    color=TEXT_COLOR,
    labelpad=8
)

ax.set_ylabel("")

# 10. NO GRID
#
# Keeps the panel visually consistent with the schematic
# rather than making it look like a standalone statistical
# plot.

ax.grid(False)

# 11. SPINES

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)

ax.spines["bottom"].set_color(
    TEXT_COLOR
)

ax.spines["bottom"].set_linewidth(
    0.7
)

# 11. PANEL FRAME

FRAME_COLOR = "#183153"   # same navy as schematic

for spine in [
    "top",
    "right",
    "bottom",
    "left"
]:
    ax.spines[spine].set_visible(True)
    ax.spines[spine].set_color(FRAME_COLOR)
    ax.spines[spine].set_linewidth(0.8)

# 12. X TICKS


ax.tick_params(
    axis="x",
    labelsize=9,
    colors=TEXT_COLOR,
    width=0.7,
    length=3
)


# 13. Y LIMITS

ax.set_ylim(
    len(family_order) - 0.5,
    -0.5
)

# 14. LAYOUT

plt.subplots_adjust(
    left=0.17,
    right=0.99,
    bottom=0.22,
    top=0.98
)

from matplotlib.lines import Line2D

present = set(vhf_df[SPECIES_COL])
legend_species = [s for s in TOP_SPECIES if s in present]   # dict order = by seroprevalence

handles = [
    Line2D([0], [0], marker="o", linestyle="none", markersize=5,
           markerfacecolor=TOP_SPECIES[s][1], markeredgecolor="none")
    for s in legend_species
]
labels = [f"{TOP_SPECIES[s][0]} ({TOP_SPECIES[s][2]:.0%})" for s in legend_species]

ax.legend(
    handles, labels,
    title="Top species",
    title_fontsize=7.5,
    fontsize=7.5,
    loc="lower right",
    frameon=False,
    handletextpad=0.3,
    labelspacing=0.35,
    borderaxespad=0.6,
)
# 15. SAVE

plt.savefig(
    "./output_figures/Fig1_VHF_peptide_reactivity_landscape.svg",
    bbox_inches="tight",
    transparent=True
)

plt.show()
```

## How does the distribution for the most seroprevalent peptides look like per family? colored by genus


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from matplotlib.lines import Line2D
from matplotlib.ticker import MultipleLocator


# =========================================================
# SETTINGS
# =========================================================

MIN_HITS = 20
BROAD_THRESHOLD = 0.20
MIN_BROAD_PEPTIDES = 2
N_COHORT = 397

OTHER_COLOR = "#c7c7c7"

GENUS_PALETTE = {
    "Mammarenavirus":  "#2166AC",
    "Phlebovirus":     "#1B9E77",
    "Orthonairovirus": "#D95F02",
    "Hartmanivirus":   "#7570B3",
    "Orthoflavivirus": "#2E8B57",
    "Uukuvirus":       "#E78AC3",
    "Bandavirus":      "#A6761D",
    "Orthohantavirus": "#666666",
}


# =========================================================
# 1. PREPARE DATA
# =========================================================

plot_df = pd.concat(
    [
        df.assign(family=family)
        for family, df in family_hits_sub_dfs.items()
    ],
    ignore_index=True
)

plot_df["sensitivity"] = pd.to_numeric(
    plot_df["sensitivity"],
    errors="coerce"
)

plot_df = plot_df.dropna(
    subset=["peptide", "family", "genus", "sensitivity"]
).copy()


# =========================================================
# 2. REMOVE NON-VIRAL FAMILIES
# =========================================================

NON_VIRAL_FAMILIES = [
    "Trypanosomatidae",
]

plot_df = plot_df[
    ~plot_df["family"].isin(NON_VIRAL_FAMILIES)
].copy()


# =========================================================
# 3. KEEP FAMILIES WITH >= MIN_HITS UNIQUE PEPTIDES
# =========================================================

family_counts = (
    plot_df
    .groupby("family")["peptide"]
    .nunique()
)

families_keep = family_counts[
    family_counts >= MIN_HITS
].index

plot_df = plot_df[
    plot_df["family"].isin(families_keep)
].copy()


# =========================================================
# 4. IDENTIFY GENERA WITH RECURRENT BROAD REACTIVITY
#
# Highlight only genera containing >=2 unique peptides
# individually reactive in >=20% of the cohort.
# =========================================================

genus_broad_summary = (
    plot_df.loc[
        plot_df["sensitivity"] >= BROAD_THRESHOLD
    ]
    .groupby("genus")
    .agg(
        n_broad_peptides=("peptide", "nunique"),
        max_sensitivity=("sensitivity", "max")
    )
    .sort_values(
        ["n_broad_peptides", "max_sensitivity"],
        ascending=[False, False]
    )
)

highlight_genera = (
    genus_broad_summary.loc[
        genus_broad_summary["n_broad_peptides"]
        >= MIN_BROAD_PEPTIDES
    ]
    .index
    .tolist()
)


# Unresolved / legacy taxonomic labels remain gray
UNRESOLVED_GENERA = [
    "Arenavirus",
]

highlight_genera = [
    genus
    for genus in highlight_genera
    if genus not in UNRESOLVED_GENERA
]


print("Genus-level broad reactivity summary:")
print(genus_broad_summary)

print("\nGenera highlighted:")
print(highlight_genera)


# =========================================================
# 5. CHECK THAT ALL HIGHLIGHTED GENERA HAVE COLORS
# =========================================================

missing_genera = [
    genus
    for genus in highlight_genera
    if genus not in GENUS_PALETTE
]

if missing_genera:
    raise ValueError(
        f"Missing colors for highlighted genera: {missing_genera}"
    )


# =========================================================
# 6. FAMILY SUMMARY
#
# Order families by their most broadly recognized peptide.
# =========================================================

family_summary = (
    plot_df
    .groupby("family")
    .agg(
        n_peptides=("peptide", "nunique"),
        max_sensitivity=("sensitivity", "max"),
        median_sensitivity=("sensitivity", "median")
    )
    .sort_values(
        "max_sensitivity",
        ascending=False
    )
)

family_order = family_summary.index.tolist()


# Highest maximum prevalence at top
y_positions = {
    family: len(family_order) - 1 - i
    for i, family in enumerate(family_order)
}


# =========================================================
# 7. FIGURE
# =========================================================

fig, ax = plt.subplots(
    figsize=(9.5, 5.8)
)

rng = np.random.default_rng(42)


# =========================================================
# 8. PLOT PEPTIDES
# =========================================================

for family in family_order:

    ypos = y_positions[family]

    sub = plot_df[
        plot_df["family"] == family
    ].copy()

    # restrained vertical jitter
    y = ypos + rng.uniform(
        -0.20,
        0.20,
        size=len(sub)
    )


    # -----------------------------------------------------
    # Define categories
    # -----------------------------------------------------

    is_highlight_genus = sub["genus"].isin(
        highlight_genera
    )

    is_broad = (
        sub["sensitivity"] >= BROAD_THRESHOLD
    )


    # -----------------------------------------------------
    # A. OTHER GENERA
    # small + gray
    # -----------------------------------------------------

    mask = ~is_highlight_genus

    ax.scatter(
        sub.loc[mask, "sensitivity"],
        y[mask.to_numpy()],
        s=11,
        color=OTHER_COLOR,
        alpha=0.50,
        edgecolor="none",
        rasterized=True,
        zorder=2
    )


    # -----------------------------------------------------
    # B. HIGHLIGHTED GENERA, BUT <20%
    # colored + small
    # -----------------------------------------------------

    mask = (
        is_highlight_genus
        & ~is_broad
    )

    colors = [
        GENUS_PALETTE[g]
        for g in sub.loc[mask, "genus"]
    ]

    ax.scatter(
        sub.loc[mask, "sensitivity"],
        y[mask.to_numpy()],
        c=colors,
        s=12,
        alpha=0.60,
        edgecolor="none",
        rasterized=True,
        zorder=3
    )


    # -----------------------------------------------------
    # C. HIGHLIGHTED GENERA AND >=20%
    # colored + large
    # -----------------------------------------------------

    mask = (
        is_highlight_genus
        & is_broad
    )

    colors = [
        GENUS_PALETTE[g]
        for g in sub.loc[mask, "genus"]
    ]

    ax.scatter(
        sub.loc[mask, "sensitivity"],
        y[mask.to_numpy()],
        c=colors,
        s=30,
        alpha=0.95,
        edgecolor="white",
        linewidth=0.35,
        rasterized=True,
        zorder=4
    )


    # -----------------------------------------------------
    # Family median
    # -----------------------------------------------------

    median = sub["sensitivity"].median()

    ax.plot(
        [median, median],
        [ypos - 0.25, ypos + 0.25],
        color="black",
        linewidth=1.6,
        solid_capstyle="butt",
        zorder=5
    )


# =========================================================
# 9. 20% REACTIVITY THRESHOLD
# =========================================================

ax.axvline(
    BROAD_THRESHOLD,
    color="#555555",
    linestyle=(0, (3, 2)),
    linewidth=1.0,
    zorder=1
)


# =========================================================
# 10. FAMILY LABELS
#
# Family = black
# peptide count = gray
# =========================================================

yticks = [
    y_positions[family]
    for family in family_order
]

ax.set_yticks(yticks)

ax.set_yticklabels(
    family_order,
    fontsize=10.5,
    color="black"
)

for label in ax.get_yticklabels():
    label.set_horizontalalignment("right")


# peptide counts
for family in family_order:

    ypos = y_positions[family]

    n_peptides = int(
        family_summary.loc[
            family,
            "n_peptides"
        ]
    )

    ax.annotate(
        f"({n_peptides} peptides)",
        xy=(0, ypos),
        xycoords=(
            ax.get_yaxis_transform(),
            "data"
        ),
        xytext=(-8, -11),
        textcoords="offset points",
        ha="right",
        va="center",
        fontsize=7.7,
        color="#808080",
        annotation_clip=False
    )


# =========================================================
# 11. X AXIS
# =========================================================

xmax = max(
    plot_df["sensitivity"].max() * 1.04,
    0.30
)

ax.set_xlim(
    0,
    xmax
)

ax.xaxis.set_major_locator(
    MultipleLocator(0.1)
)

ax.set_xlabel(
    f"Fraction of cohort reactive to peptide (n = {N_COHORT})",
    fontsize=9.5,
    color="black",
    labelpad=7
)

ax.set_ylabel("")


# =========================================================
# 12. GRID
# =========================================================

ax.grid(
    axis="x",
    color="#e6e6e6",
    linewidth=0.65
)

ax.set_axisbelow(True)


# =========================================================
# 13. AXES
# =========================================================

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.spines["left"].set_color("#4d4d4d")
ax.spines["bottom"].set_color("#4d4d4d")

ax.spines["left"].set_linewidth(0.7)
ax.spines["bottom"].set_linewidth(0.7)

ax.tick_params(
    axis="x",
    labelsize=8.5,
    colors="#333333",
    width=0.7,
    length=3.5
)

ax.tick_params(
    axis="y",
    colors="black",
    width=0.7,
    length=4,
    pad=8
)


# =========================================================
# 14. Y LIMITS
# =========================================================

top_y = max(
    y_positions.values()
)

ax.set_ylim(
    -0.55,
    top_y + 0.55
)


# =========================================================
# 15. GENUS LEGEND
# =========================================================

genus_handles = [
    Line2D(
        [0], [0],
        marker="o",
        linestyle="none",
        markersize=6.5,
        markerfacecolor=GENUS_PALETTE[genus],
        markeredgecolor="none",
        label=genus
    )
    for genus in highlight_genera
]

# gray category
genus_handles.append(
    Line2D(
        [0], [0],
        marker="o",
        linestyle="none",
        markersize=6.5,
        markerfacecolor=OTHER_COLOR,
        markeredgecolor="none",
        label="Other genera"
    )
)


genus_legend = ax.legend(
    handles=genus_handles,
    title="Genus",
    frameon=False,

    # position
    bbox_to_anchor=(1.06, 1.0),
    loc="upper left",
    borderaxespad=0,

    # typography
    fontsize=9,
    title_fontsize=10,

    # spacing
    handletextpad=0.65,
    labelspacing=0.55,
    borderpad=0,
    handlelength=1.0
)

genus_legend.get_title().set_fontweight(
    "semibold"
)

ax.add_artist(genus_legend)


# =========================================================
# 16. FAMILY MEDIAN LEGEND
# =========================================================

median_handle = Line2D(
    [0], [0],
    color="black",
    linewidth=1.8,
    label="Family median"
)

ax.legend(
    handles=[median_handle],
    frameon=False,

    # aligned with genus legend
    bbox_to_anchor=(1.06, 0.53),
    loc="upper left",
    borderaxespad=0,

    fontsize=9,
    handlelength=1.8,
    handletextpad=0.65
)


# =========================================================
# 17. LAYOUT
# =========================================================

plt.subplots_adjust(
    left=0.20,
    right=0.72,
    bottom=0.14,
    top=0.97
)


# =========================================================
# 18. SAVE
# =========================================================

"""plt.savefig(
    "viral_family_peptide_reactivity_by_cohort_prevalence.svg",
    format="svg",
    bbox_inches="tight"
)

plt.savefig(
    "viral_family_peptide_reactivity_by_cohort_prevalence.pdf",
    format="pdf",
    bbox_inches="tight"
)
"""
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

hits = pep_hits_meta.reset_index()

# ---------------------------------------------------------
# Consistent genus colors across figures
# ---------------------------------------------------------
GENUS_PALETTE = {
    'Orthonairovirus': '#2d6a8e',
    'Phlebovirus':     '#b4432f',
    'Mammarenavirus':  '#6a8e2d',
    'Orthohantavirus': '#8e6a2d',
    'Uukuvirus':       '#5a4a7a',
    'Orthoflavivirus': '#2d7a6a',
}

DEFAULT_COLOR = '0.70'


def plot_top_genera_grouped(hits, top_k=8, fname=None):

    # ---------------------------------------------------------
    # 1. Count unique reactive peptides per genus
    # ---------------------------------------------------------
    g = (
        hits.groupby(["family", "genus"])["peptide"]
            .nunique()
            .reset_index(name="n")
            .sort_values("n", ascending=False)
    )

    # Select top genera OVERALL
    top = g.head(top_k).copy()

    # ---------------------------------------------------------
    # 2. Order families by strongest genus
    # ---------------------------------------------------------
    family_order = (
        top.groupby("family")["n"]
           .max()
           .sort_values(ascending=False)
           .index
           .tolist()
    )

    # ---------------------------------------------------------
    # 3. Build plotting order + gaps between families
    # ---------------------------------------------------------
    rows = []
    y_positions = []
    separators = []

    y = 0
    gap = 0.65

    for i, fam in enumerate(family_order):

        fam_df = (
            top[top["family"] == fam]
            .sort_values("n", ascending=False)
        )

        for _, row in fam_df.iterrows():
            rows.append(row)
            y_positions.append(y)
            y += 1

        # Add separator/gap between families
        if i < len(family_order) - 1:
            separators.append(y - 0.5 + gap / 2)
            y += gap

    plot_df = pd.DataFrame(rows)

    # ---------------------------------------------------------
    # 4. Plot
    # ---------------------------------------------------------
    fig, ax = plt.subplots(
        figsize=(6, 5)
    )

    ax.barh(
        y_positions,
        plot_df["n"],
        height=0.72,
        color=[
            GENUS_PALETTE.get(gen, DEFAULT_COLOR)
            for gen in plot_df["genus"]
        ],
        edgecolor="none"
    )

    # Genus labels
    ax.set_yticks(y_positions)
    ax.set_yticklabels(
        plot_df["genus"],
        fontsize=11
    )

    # Largest family at top
    ax.invert_yaxis()

    # ---------------------------------------------------------
    # 5. Add counts
    # ---------------------------------------------------------
    xmax = plot_df["n"].max()

    for yi, n in zip(y_positions, plot_df["n"]):
        ax.text(
            n + xmax * 0.012,
            yi,
            f"{n}",
            va="center",
            ha="left",
            fontsize=10
        )

    ax.set_xlim(0, xmax * 1.12)

    # ---------------------------------------------------------
    # 6. Family separators
    # ---------------------------------------------------------
    for sep in separators:
        ax.axhline(
            sep,
            color="0.82",
            linewidth=1,
            linestyle="--",
            zorder=0
        )

    # ---------------------------------------------------------
    # 7. Labels / formatting
    # ---------------------------------------------------------
    ax.set_xlabel(
        "Number of reactive peptides",
        fontsize=12
    )

    ax.set_title(
        "Viral genera with the most reactive peptides",
        fontsize=14,
        pad=15
    )

    ax.spines[["top", "right", "left"]].set_visible(False)

    ax.tick_params(
        axis="y",
        length=0,
        pad=8
    )

    ax.grid(
        axis="x",
        color="0.92",
        linewidth=0.8
    )

    ax.set_axisbelow(True)

    plt.tight_layout()

    if fname:
        plt.savefig(
            fname,
            bbox_inches="tight",
            transparent=True
        )

    plt.show()

    return plot_df


top_genera = plot_top_genera_grouped(
    hits,
    top_k=6,
    fname="svg_figures/bar_top_genera_grouped.svg"
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Patch

hits = pep_hits_meta.reset_index()

def plot_top_species_by_genus(
    hits,
    genera=("Mammarenavirus", "Orthonairovirus", "Orthoflavivirus"),
    top_k=2,
    fname=None
):

    # ---------------------------------------------------------
    # 1. Count unique reactive peptides per species
    # ---------------------------------------------------------
    g = (
        hits[hits["genus"].isin(genera)]
        .groupby(["family", "genus", "species"])["peptide"]
        .nunique()
        .reset_index(name="n")
    )

    # ---------------------------------------------------------
    # 2. Select top 2 species WITHIN each genus
    # ---------------------------------------------------------
    top = (
        g.sort_values("n", ascending=False)
         .groupby("genus", group_keys=False)
         .head(top_k)
         .copy()
    )

    # Keep genus order exactly as specified above
    genus_order = list(genera)

    # ---------------------------------------------------------
    # 3. SAME COLORS as previous figure
    # ---------------------------------------------------------
    genus_colors = GENUS_COLORS

    # ---------------------------------------------------------
    # 4. Build plotting positions with gaps between genera
    # ---------------------------------------------------------
    rows = []
    y_positions = []
    separators = []

    y = 0
    gap = 0.8

    for i, genus in enumerate(genus_order):

        genus_df = (
            top[top["genus"] == genus]
            .sort_values("n", ascending=False)
        )

        for _, row in genus_df.iterrows():
            rows.append(row)
            y_positions.append(y)
            y += 1

        # Add gap between genera
        if i < len(genus_order) - 1:
            separators.append(y - 0.5 + gap / 2)
            y += gap

    plot_df = pd.DataFrame(rows)

    # ---------------------------------------------------------
    # 5. Plot
    # ---------------------------------------------------------
    fig, ax = plt.subplots(
        figsize=(13, 7)
    )

    ax.barh(
        y_positions,
        plot_df["n"],
        height=0.7,
        color=[
            genus_colors[g]
            for g in plot_df["genus"]
        ],
        edgecolor="none"
    )

    # Species labels
    ax.set_yticks(y_positions)
    ax.set_yticklabels(
        plot_df["species"],
        fontsize=17
    )

    # Largest species at top
    ax.invert_yaxis()

    # ---------------------------------------------------------
    # 6. Add counts
    # ---------------------------------------------------------
    xmax = plot_df["n"].max()

    for yi, n in zip(y_positions, plot_df["n"]):
        ax.text(
            n + xmax * 0.015,
            yi,
            f"{n}",
            va="center",
            ha="left",
            fontsize=14
        )

    ax.set_xlim(0, xmax * 1.15)

    # ---------------------------------------------------------
    # 7. Separators between genera
    # ---------------------------------------------------------
    for sep in separators:
        ax.axhline(
            sep,
            color="0.82",
            linewidth=1.2,
            linestyle="--",
            zorder=0
        )

    # ---------------------------------------------------------
    # 8. Formatting
    # ---------------------------------------------------------
    ax.set_xlabel(
        "Number of reactive peptides",
        fontsize=16,
        labelpad=8
    )

    ax.set_title(
        "Most reactive species within selected viral genera",
        fontsize=18,
        pad=15
    )

    ax.tick_params(
        axis="x",
        labelsize=14
    )

    ax.tick_params(
        axis="y",
        length=0,
        pad=8
    )

    ax.spines[["top", "right", "left"]].set_visible(False)

    # Subtle x-axis grid
    ax.grid(
        axis="x",
        color="0.92",
        linewidth=0.8
    )

    ax.set_axisbelow(True)

    # ---------------------------------------------------------
    # 9. Legend = genus
    # ---------------------------------------------------------
    handles = [
        Patch(
            facecolor=genus_colors[genus],
            label=genus
        )
        for genus in genus_order
    ]

    ax.legend(
        handles=handles,
        title="Genus",
        frameon=False,
        bbox_to_anchor=(1.02, 1),
        loc="upper left",
        fontsize=13,
        title_fontsize=13
    )

    plt.tight_layout()

    if fname:
        plt.savefig(
            fname,
            bbox_inches="tight",
            transparent=True
        )

    plt.show()

    return plot_df


top_species = plot_top_species_by_genus(
    hits,
    genera=[
        "Mammarenavirus",
        "Orthonairovirus",
        "Orthoflavivirus"
    ],
    top_k=2,
    fname="svg_figures/bar_top_species_selected_genera.svg"
)
```

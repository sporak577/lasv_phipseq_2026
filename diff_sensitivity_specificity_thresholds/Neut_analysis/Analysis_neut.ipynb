## The objective of this analysis is to identify the dominant peptides in individuals with high neutralization vs moderate vs low neutralization. I am particularly interested to see if seroreactivity to tile position 19, which is the top enriched peptide of monoclonal antibodies 37.7H and 25.6A by PhIP-Seq (both neutralizing monoconal antibodies of the GPC-B epitope group) are dominant targets of the polyclonal response in neutralizing individuals. 

```python
import pandas as pd

meta_neut = pd.read_csv('../files/LASV_ELISA_Neut_and_PhIP_Seq_result_corrected_w_month_year_filtered_first_tp.csv')

meta_peps = pd.read_csv('../files/lassa_library_sequences_species_standardized_070726_ICTV_corrected.csv')

filtered_peps = pd.read_csv('../files/pep_hits_sen_spec_092326.csv')

long = pd.read_csv('../files/long_092326_senspec.csv')
```

```python
meta_neut = meta_neut.drop(columns=['Unnamed: 0'])

filtered_peps = filtered_peps.drop(columns=['Unnamed: 0'])
```

```python
meta_neut = meta_neut[['individual', 'NP_FINAL_RESULT', 'GP_FINAL_RESULT', 'Neut_Percent', 'Neut_Result ', 'Serology_Result_Corrected']]
```

## merge the neutralization info to the long dataframe on individual

```python
long_neut = long.merge(meta_neut, on='individual', how='left')
```

```python
long_neut.columns
```

## subset to individuals which are neutralizing

```python
long_neut['Neut_Result '].unique()
```

```python
neut_levels = ['Strong', 'Moderate', 'Weak']

neut_samples = long_neut.loc[long_neut['Neut_Result '].isin(neut_levels)]

non_neut_samples =  long_neut.loc[long_neut['Neut_Result '] == 'Non-Neutralizing']
```

## save to csv

```python
neut_samples.to_csv('long_neut_samples_092326.csv', index=False)

non_neut_samples.to_csv('long_non_neut_samples_092326.csv', index=False)
```

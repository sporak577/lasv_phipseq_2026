```python
metadata = pd.read_csv('master_sample_metadata_060626.csv')
```

### subset to the first timepoint for each individual

```python
# For each individual (ID), keep only the row with the earliest Date_text
metadata_earliest = (
    metadata
    .dropna(subset=['sample_id', 'participant_id', 'collection_year', 'collection_month'])
    .sort_values(['sample_id', 'participant_id', 'collection_year', 'collection_month'])
    .drop_duplicates('participant_id', keep='first')
)

print(f'Original metadata rows: {len(metadata)}')
print(f'After subsetting to earliest timepoint: {len(metadata_earliest)}')
metadata_earliest[['sample_id', 'participant_id', 'collection_year', 'collection_month']].head(10)
```

```python
metadata_earliest_survivors = metadata_earliest[metadata_earliest['participant_id'].str.upper().str.startswith('S')]

print(f'Earliest timepoint (all): {len(metadata_earliest)}')
print(f'Earliest timepoint (survivors only): {len(metadata_earliest_survivors)}')
metadata_earliest_survivors[['sample_id', 'participant_id', 'collection_year', 'collection_month']].head(10)
```

```python
earliest_sample_ids = set(metadata_earliest['sample_id'])
keep_cols = [c for c in df.columns if c in earliest_sample_ids]
earliest_tp_df = df[keep_cols]

print(f'df columns: {df.shape[1]}, matched to earliest timepoint: {len(keep_cols)}')
earliest_tp_df.to_csv('earliest_tp_rpk_avg_value_for_cohort_techreps_071626.csv')
```

```python
df = pd.read_csv('long_091826.csv')
```

```python
df = df.loc[df['cohort'] == 'SL']
```

```python
sum_species_df = df.groupby(['species', 'individual'])['reactive'].sum().sort_values(ascending=False)
```

```python
sum_species_df =sum_species_df.reset_index()
```

```python
sum_species_df.to_csv('sum_reactive_peps_species_per_ind.csv', index=False)
```
